```python
with open("sample.txt", "r") as file:
    print(file.read())     # to open and close file without using close()
```

```python
with open("sample.txt", "a") as file:    #r+, rb, w+, 
    print(file.write("\nline 3"))     # to open and close file without using close()
```

```python
with open("sample.txt", "r") as file:
    print(file.readlines())     # to open and close file without using close()
```

```python
# tell current position of pointer in the file
file = open("sample.txt", "r")

print(file.tell())

file.close()
```

```python
file = open("sample.txt", "r")
file.read(2)

print(file.tell())
file.close()
```

```python
file = open("sample.txt", "r")

file.seek(0)

print(file.read())

file.close()
```

```python
file = open("sample.txt", "r")      # open the file in read mode

file.read(7)      # read first 7 characters

print(file.tell())     # to tell the current position of pointer

file.seek(0)            # to get current position of pointer

print(file.tell())

file.close()
```

```python
# readlines and writelines
languages = ["Python\n", "Java\n", "C++"]

file = open("sample4.txt", "w")

file.writelines(languages)

file.close()
```

```python
fruits = ["Apple\n", "Banana\n", "Orange"]

file = open("sample5.txt", "w")

file.writelines(fruits)

file.close()
```
