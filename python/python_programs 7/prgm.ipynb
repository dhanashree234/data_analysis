Take a number N and print an N × N square of stars.

```python
N = int(input("Enter a number: "))

for i in range(N):          # outer loop → rows
    for j in range(N):      # inner loop → columns
        print("*", end="")
    print()                 # move to next line
```

```python
N = int(input("Enter a number: "))

for _ in range(N):
    print("*" * N)
```

Nested list

```python
numbers = [
    [10, 20],
    [30, 40],
    [50, 60]
]

total = 0

for row in numbers:

    for number in row:
        total += number

print("Total:", total)
```

Count all even numbers in a nested list.

```python
nested = [[1, 2, 3], [4, 5, 6], [7, 8]]

count = 0

for sublist in nested:
    for num in sublist:
        if num % 2 == 0:
            count += 1

print("Total even numbers:", count)
```

Create a list of programming languages. Take a language from the user and search for it. Stop the loop when it is found.

```python
languages = ["Python", "Java", "C++", "JavaScript", "Go", "Rust", "Ruby"]

user_lang = input("Enter a programming language: ")

for lang in languages:
    if lang.lower() == user_lang.lower():
        print(f"{lang} found in the list.")
        break
else:
    print("Language not found.")
```

Print numbers from a list. Stop processing when a negative number is found.

```python
numbers = [10, 5, 3, 0, -2, 7, 9]

for n in numbers:
    if n < 0:
        print("Negative number found. Stopping.")
        break
    print(n)
```

Allow the user to enter a password a maximum of 3 times. Stop the loop immediately when the correct password is entered.

```python
correct_password = "secret123"

for attempt in range(3):
    pwd = input("Enter password: ")
    if pwd == correct_password:
        print("login successful.")
        break
    else:
        print("Incorrect password.")
else:
    print("maximum attempts reached.")
```

list and list comprehension

```python
numbers = []

for i in range(1, 6):
    numbers.append(i)

print(numbers)
```

```python
# list comprehension : we give condition in this
numbers = [i for i in range(1, 6)]

print(numbers)
```

```python
result = ["Even" if i % 2 == 0 else "Odd" for i in range(1, 6)]

print(result)
```

```python
# positional args
def add(a, b):
    print(a + b)

add(10, 20)

# variable-length args
# *args
# **kwargs
```

```python
def display(*numbers):
    print(numbers)

display(10, 20, 30, 40)
```

```python
def add(*numbers):
    total = 0

    for i in numbers:
        total = total + i

    print("Sum:", total)


add(10, 20)
add(10, 20, 30)
add(10, 20, 30, 40)

# positional args
```

```python
# kwargs

def student(**details):
    print(details)

student(name="Arun", age=20, course="Python")
```
