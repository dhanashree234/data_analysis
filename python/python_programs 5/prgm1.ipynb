1. Take a number. First check whether it is positive. If it is positive, check whether it is even or odd. Otherwise, display "Number is Zero or Negative".

```python
class Number:
    def accept(self):
        self.num = int(input("Enter first number:"))
    
    def check(self):
        if self.num > 0:
            if self.num % 2 == 0:
                return "Number is positive and even"
            else:
                return "Number is positive and odd"
        elif self.num < 0:
            return "Number is negative"
        else:
            return "Number is zero"
    
    def display(self):
        print(self.check())       
```

```python
# Create object
obj = Number()
```

```python
# Call methods
obj.accept()
```

```python
obj.display()
```
