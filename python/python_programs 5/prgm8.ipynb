8.	A student must score at least 60 marks to apply for admission. If the marks requirement is met, check whether the student passed the entrance exam.

```python
class Admission:
    def accept(self):
        self.marks = int(input("Enter marks: "))
        self.exam = input("Passed entrance exam? (yes/no): ").lower()

    def check(self):
        if self.marks >= 60:
            if self.exam == "yes":
                return "Eligible for admission."
            else:
                return "Entrance exam not passed."
        else:
            return "Marks requirement not met."

    def display(self):
        print(self.check())
```

```python
obj = Admission()
```

```python
obj.accept()
```

```python
obj.display()
```
