10.	A customer can place an order if the product is available. If available, check whether the customer has sufficient balance to pay for the product.

```python
class Order:
    def accept(self):
        self.available = input("Is product available? (yes/no): ").lower()
        self.balance = float(input("Enter customer balance (£): "))
        self.price = float(input("Enter product price (£): "))

    def check(self):
        if self.available == "yes":
            if self.balance >= self.price:
                return "Order Placed Successfully."
            else:
                return "Insufficient Balance."
        else:
            return "Product Not Available."

    def display(self):
        print(self.check())
```

```python
obj = Order()
```

```python
obj.accept()
```

```python
obj.display()
```
