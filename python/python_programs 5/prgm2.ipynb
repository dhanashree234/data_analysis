2.	Take a username and password. First check whether the username is "admin". If correct, check whether the password is "python123". Display an appropriate message for every result.

```python
class Account:
    def accept(self):
        self.username = input("Enter username:")
        self.password = input("Enter password:")
    
    def check(self):
        if self.username =="admin" and self.password == "python123":
            return "Login successful"
        else:
            return "Login unsuccessful"
    
    def display(self):
        print(self.check())       
```

```python
# Create object
obj = Account()
```

```python
# Call methods
obj.accept()
```

```python
obj.display()
```
