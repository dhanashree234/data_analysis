6.	 An employee is eligible for a service bonus if they have worked for at least 5 years. If eligible, check whether their performance rating is at least 8 for an additional performance bonus.


```python
class Employee:
    def accept(self):
        self.years = int(input("Enter years of service: "))
        self.rating = float(input("Enter performance rating: "))

    def check(self):
        if self.years >= 5:
            if self.rating >= 8:
                return "Eligible for service and performance bonus."
            else:
                return "Eligible for service bonus only."
        else:
            return "Not eligible for bonus."

    def display(self):
        print(self.check())
```

```python
obj = Employee()
```

```python
obj.accept()
```

```python
obj.display()
```
