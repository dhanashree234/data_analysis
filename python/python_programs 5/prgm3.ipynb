3.	Take a number. First check whether it is divisible by 2. If yes, check whether it is also divisible by 5. Display an appropriate message for every result.

```python
class Number:
    def accept(self):
        self.num = int(input("Enter first number:"))
    
    def check(self):
        if self.num % 2 == 0:
            if self.num % 5 == 0:
                return "Number is divisible by 2 and 5 both"
        else:
                return "Number is not divisible by 2 and 5 both"
    
    def display(self):
        print(self.check())       
```

```python
# Create object
obj = Number()
```

```python
# Call methods
obj.accept()
```

```python
obj.display()
```
