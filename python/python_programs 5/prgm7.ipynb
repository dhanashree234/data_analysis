7.	Take the user's PIN. If the PIN is correct, check whether the withdrawal amount is within the available balance. Display an appropriate message for every result.

```python
class ATM:
    def accept(self):
        self.pin = input("Enter PIN: ")
        self.balance = 5000
        self.amount = float(input("Enter withdrawal amount: "))

    def check(self):
        if self.pin == "1234":
            if self.amount <= self.balance:
                return "Transaction Successful."
            else:
                return "Insufficient Balance."
        else:
            return "Incorrect PIN."

    def display(self):
        print(self.check())
```

```python
obj = ATM()
```

```python
obj.accept()
```

```python
obj.display()
```
