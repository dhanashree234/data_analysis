```python
import math
def round_up(num, decimals=0):
   multiplier = 10 ** decimals
   return math.ceil(num * multiplier) / multiplier
print(round_up(2.1)) # 3.0
print(round_up(2.23, 1)) # 2.3
print(round_up(2.543, 2)) # 2.55
```
