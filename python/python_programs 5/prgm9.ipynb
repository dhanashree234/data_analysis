9.	A customer must earn at least £2,000 per month to apply for a loan. If the salary requirement is met, check whether their credit score is at least 700.

```python
class Loan:
    def accept(self):
        self.salary = float(input("Enter monthly salary (£): "))
        self.credit = int(input("Enter credit score: "))

    def check(self):
        if self.salary >= 2000:
            if self.credit >= 700:
                return "Loan Approved."
            else:
                return "Credit score too low."
        else:
            return "Salary requirement not met."

    def display(self):
        print(self.check())
```

```python
obj = Loan()
```

```python
obj.accept()
```

```python
obj.display()
```
