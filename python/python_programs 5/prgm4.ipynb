4.	Take a number. First check whether it is greater than or equal to 1. If yes, check whether it is less than or equal to 100. Display an appropriate message.

```python
class Number:
    def accept(self):
        self.num = int(input("Enter first number:"))
    
    def check(self):
        if self.num >= 1:
            if self.num <= 100:
                return "Number is positive and even"
            else:
                return "Number is positive and odd"
        elif self.num < 0:
            return "Number is negative"
        else:
            return "Number is zero"
    
    def display(self):
        print(self.check())       
```

```python
# Create object
obj = Number()
```

```python
# Call methods
obj.accept()
```

```python
obj.display()
```
