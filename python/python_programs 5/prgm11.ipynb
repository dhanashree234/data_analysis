10.	A customer can place an order if the product is available. If available, check whether the customer has sufficient balance to pay for the product.

```python
class Course:
    def accept(self):
        self.basic = input("Completed Basic Python course? (yes/no): ").lower()
        self.assessment = input("Passed final assessment? (yes/no): ").lower()

    def check(self):
        if self.basic == "yes":
            if self.assessment == "yes":
                return "Eligible for Advanced Python course."
            else:
                return "Final assessment not passed."
        else:
            return "Complete the Basic Python course first."

    def display(self):
        print(self.check())
```

```python
obj = Course()
```

```python
obj.accept()
```

```python
obj.display()
```
