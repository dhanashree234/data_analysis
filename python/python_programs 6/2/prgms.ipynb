3. Print multiplication tables from 1 to 5 using nested loops.

```python
for i in range(1,6):
    print(f"\nMultiplication table of {i}\n")
    for j in range(1,11):
        print(f"{i} x {j}  =  ", i * j)
```

```python
# 10. Create a set containing five different numbers and print it.
#     Add an element to a set.
#     Remove an element from a set.
#     Check whether a particular value exists in a set.

# Creating a set
numbers = {10, 20, 30, 40, 50}
print(numbers)

# Adding an element
numbers.add(60)
print("After adding 60:", numbers)

# Removing an element
numbers.remove(30)
print("After removing 30:", numbers)

# Checking whether a value exists
if 40 in numbers:
    print("40 exists in the set")
else:
    print("40 does not exist in the set")
```
