Print numbers from 1 to 10.

```python
i = 1

while(i<=10):
    print(i)
    i += 1
```

```python
for i in range(1,11):
    print(f"{i}")
```

```python
print(*range(1, 11), sep="\t")              # * (Unpacking Operator) : The * unpacks the values from the range object into 
                                            # separate arguments.
```

Print numbers from 10 to 1.

```python
print(*range(10, 0, -1), sep="\t")                  # range(start, stop, step(here decrease by -1))
```

Print all even numbers from 1 to 20.


```python
print(*range(2,11,2), sep="\t")
```

Print all odd numbers from 1 to 20.

```python
print(*range(1,11,2))
```

Print the multiplication table of 5.


```python
print(*[f"5 x {i} = {5 * i}" for i in range(1, 11)], sep="\n")
```

Take a number from the user and print its multiplication table.

```python
num = int(input("Enter number: "))

print(*[f"{num} x {i} = {num * i}" for i in range(1,11)], sep="\n")
```

Find the sum of numbers from 1 to 10.

```python
del sum                     # to avoid TypeError: 'int' object is not callable error
print(sum(range(1, 11)))
```

```python
sum = 0
for i in range(1,11):
    sum+=i
print(sum)
```
