```python
class BankAccount:
  def __init__(self, account_name, account_number, balance=0):
    self.account_name = account_name
    self.account_number = account_number
    self.balance = balance

  def deposit(self, amount):
    if amount > 0:
      self.balance += amount
      print(f"Deposited ${amount}) New balance: ${self.balance}")
    else:
      print("Invalid deposit amount. Amount must be greater than 0.")

  def withdraw(self, amount):
    if amount <= self.balance:
      self.balance -= amount
      print(f"Withdrew ${amount}. New balance: ${self.balance}")
    else:
      print("Invalid withdrawal amount or insufficient funds.")

  def get_balance(self):
    return self.balance

  def account_info(self):
    print(f"Account Name: {self.account_name}")
    print(f"Account Number: {self.account_number}")
    print(f"Balance: ${self.balance}")
```

```python
class SavingsAccount(BankAccount):
  def __init__(self, account_name, account_number, balance, interest_rate):
    super().__init__(account_name, account_number, balance)
    self.interest_rate = interest_rate

  def account_info(self):
    super().account_info()
    print(f"interest rate = {self.interest_rate * 100}")
```

```python
s = SavingsAccount("aname",100,1000,5)
s.deposit(200)
s.withdraw(100)
s.account_info()
```

```python
class Ticket:
  def __init__(self, ticket_id, movie_name, seat_no, price):
    self.ticket_id = ticket_id
    self.movie_name = movie_name
    self.seat_no = seat_no
    self.price = price

  def display_ticket(self):
    print(f"Ticket ID: {self.ticket_id}")
    print(f"Movie Name: {self.movie_name}")
    print(f"Seat No: {self.seat_no}")
    print(f"Price: {self.price}")
```

```python
class TicketBookingSystem:
  def __init__(self):
    self.tickets = [] #empty list to store booked tickets

  def book_ticket(self, movie_name, seat_no, price):
    ticket_id = len(self.tickets) + 1
    ticket = Ticket(ticket_id, movie_name, seat_no, price)
    self.tickets.append(ticket)
    print(f"Ticket booked - {ticket_id}")

  def display_ticket(self):
    if not self.tickets:
      print("No ticket booked")
    else:
      print("All booked")
      for ticket in self.tickets:
        ticket.display_ticket()

  def cancel_ticket(self, ticket_id):
    for ticket in self.tickets:
      if ticket.ticket_id == ticket_id:
        self.tickets.remove(ticket)
        print(f"{ticket_id} cancelled.")
        return
      print("Not found")
```

```python
t = TicketBookingSystem()
t.book_ticket("movie1","n1",50)
t.book_ticket("movie2","n1",50)
t.book_ticket("movie3","n1",50)

print("\nDisplay tickets")
t.display_ticket()

print("\nCancelled tickets")
t.cancel_ticket(2)

print("\nDisplay tickets")
t.display_ticket()
```
