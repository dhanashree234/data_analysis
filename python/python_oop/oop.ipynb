```python
# Define the Student class
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
```

```python
# Create 3 student objects
student1 = Student("Alice", 85)
student2 = Student("Bob", 90)
student3 = Student("Charlie", 78)
```

```python
# Store students in a list
students = [student1, student2, student3]
```

```python
# Print each student's name and marks
for student in students:
    print("Name:", student.name)
    print("Marks:", student.marks)
    print()
```

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price
```

```python
p1 = Product("xyz", 100)
p1.discount = 10
f_price = p1.price - (p1.price * p1.discount)/100
print(f"Name : {p1.name}")
print(f"Original price : {p1.price}")
print(f"discount : {p1.discount}")
print(f"final price : {f_price}")
```

```python
class Engine:
    def __init__(self, horsepower):
        self.horsepower = horsepower
        
class Car:
    def __init__(self, brand, engine):
        self.brand = brand
        self.engine = engine
```

```python
# Create an Engine object
engine = Engine(150)

# Create a Car object with the Engine
car = Car("Toyota", engine)
```

```python
# Print car brand and engine horsepower
print("Car Brand:", car.brand)
print("Engine Horsepower:", car.engine.horsepower)
```

```python
class Book:
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self.price = price
        
    def __str__(self):
        return f"{self.title} {self.author} {self.price}"
```

```python
b1 = Book("zxc", "qwe", 100)
print(b1)
```

```python
class Address:
    def __init__(self, city, zip_code):
        self.city = city
        self.zip_code = zip_code
        
    def __str__(self):
        return f"{self.city} {self.zip_code}"

class Person:
    def __init__(self, name, address):
        self.name = name
        self.address = address
        
    def __str__(self):
        return f"{self.name} {self.address}"
```

```python
a1 = Address("zxc", 12345)
p1 = Person("qwe", "addr1")
```

```python
print(a1)
print(p1)
```

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
```

```python
emp = [emp1, emp2, emp3]
```

```python
# Create Circle with radius. Print radius and area in str().
class Circle:
    def __init__(self, radius):
        self.radius = radius
    
    def calculate(self):
        return 3.142 * self.radius**2
```

```python
c = Circle(10)
area = c.calculate()
print(f"Area = {area}")
```

```python
# Create Light with status (True/False). Print "Light is ON" or "Light is OFF".
class Light:
    def __init__(self, status):
        self.status = status

    def operate(self):
        if self.status == True:
            print("ON")
        else:
            print("OFF")        
```

```python
l = Light(True)
l.operate()
```
