```python
df = pd.read_csv("Noisy_Student_Bag_Price_Prediction_Dataset.csv")
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.fillna(0, inplace=True)
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum() #to check duplicates
```

```python
df.describe()
```

```python
df.info()
```

```python
# find Average and maximum price per Brand type
df.groupby('Brand')['Price'].agg(['mean','max'])

# Average price per Brand and material
df.groupby(['Brand', 'Material'])['Price'].mean()

# Add a column showing price deviation from the brand average
df['Price_Deviation'] = df.groupby('Brand')['Price'].transform(lambda x: x - x.mean())
df
```

```python
df = pd.read_csv('Sales Dataset.csv')
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
# Total Amount by Category here total amount is displayed by category
pd.pivot_table(df, 
               values='Amount', 
               index='Category', 
               aggfunc='sum')
```

```python
# Average Profit by Sub-Category 
pd.pivot_table(df,
               values='Profit',
               index='Sub-Category',
               aggfunc='mean')
```

```python
# Total Quantity Sold by Category and Sub-Category    
pd.pivot_table(df,
               values='Amount',
               index='PaymentMode',
               aggfunc='mean')
```

```python
orders_df = pd.read_csv("restaurant/orders.csv")
```

```python
orders_df
```

```python
restaurant_df = pd.read_csv("restaurant/restaurants.csv")
```

```python
restaurant_df
```

```python
members_df = pd.read_csv("restaurant/members.csv")
members_df
```

```python
# Concatenate members and restaurants side by side to explore their structure.
combined = pd.concat([members_df.head(), restaurant_df.head()], axis=1)
combined
```

```python
# Merge orders with members to see who placed each order.
orders_with_members = pd.merge(orders_df, members_df, on='id', how='inner')
orders_with_members.head(10)
```

```python
# Merge orders with restaurants to see restaurant details for each order.
orders_with_restaurants = pd.merge(orders_df,restaurant_df,on='id',how='inner')
orders_with_restaurants
```

```python
# Merge all three: orders, members, and restaurants to get a full dataset.
orders_full_dataset = pd.merge(orders_df, members_df, on='id', how='inner')
orders_full_dataset = pd.merge(orders_full_dataset, restaurant_df, on='id', how='left')
orders_full_dataset.head(11)
```

```python
# Join orders with restaurants using the index.
members_df.reset_index(inplace=True)
members_df.set_index('id', inplace=True)
orders_with_members = orders_df.join(members_df, on='id', how='left')
orders_with_members.head()
```

```python
# Join orders with members using the index.
popular_restaurants = orders_full_dataset.groupby('restaurant_name')['id'].count().reset_index()
popular_restaurants = popular_restaurants.sort_values(by='id', ascending=False)
popular_restaurants.head()
```
