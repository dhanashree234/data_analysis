**MATPLOTLIB**

* Matplotlib is a powerful Python library for data visualization.

* It is widely used for creating static, animated, and interactive plots.

Why use Matplotlib?

* To visualize datasets (instead of just numbers).

* To create different types of plots: line charts, bar charts, scatter plots, histograms, pie charts, etc.

* To customize plots (colors, labels, titles, legends, grid).

* Works well with NumPy, Pandas, and SciPy for data analysis.

Commonly Used Functions:

**plot()**: For line plots.

**scatter()**: For scatter plots.

**bar()**: For bar charts.

**hist()**: For histograms.

**pie()** : For pie charts.

**LINE PLOT**

* A line plot shows the relationship between x (independent variable) and y (dependent variable).

* Useful for trends over time (e.g., stock prices, temperature changes).

```python
import matplotlib.pyplot as plt
```

```python
x = [10, 20, 30, 40, 50]
y = [100, 200, 300, 400, 500]

plt.plot(x, y)
plt.title("Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

**BAR CHART**

* A bar chart shows comparison between categories.

* Each bar length represents the value of that category.

```python
categories = ["Math", "Science", "English", "History"]
marks = [88, 92, 80, 76]

plt.bar(categories, marks, color="skyblue")
plt.title("Student Marks in Subjects")
plt.ylabel("Marks")
plt.show()
```

**SCATTER PLOT**

* A scatter plot shows the relationship between two variables.

* Each point represents an observation.

* Used to find patterns (e.g., height vs. weight).

```python
height = [150, 160, 165, 170, 175, 180]
weight = [50, 55, 60, 65, 70, 80]

plt.scatter(height, weight, color="green")
plt.title("Height vs Weight")
plt.xlabel("Height (cm)")
plt.ylabel("Weight (kg)")
plt.show()
```

**PIE CHART**

* A pie chart represents data as a percentage of the whole.

* Each slice = proportion of category.

```python
fruits = ["Apple", "Banana", "Mango", "Grapes"]
values = [30, 25, 20, 25]

plt.pie(values, labels=fruits, autopct="%1.1f%%", startangle=90)
plt.title("Fruit Distribution")
plt.show()
```

**HISTOGRAM**

* A histogram shows how data is distributed.

* Data is divided into bins (intervals).

Example: Student marks distribution.

```python
import numpy as np

data = np.random.randint(50, 100, 100)  # 100 random marks between 50-100

plt.hist(data, bins=10, color="purple", edgecolor="black")
plt.title("Marks Distribution")
plt.xlabel("Marks Range")
plt.ylabel("Number of Students")
plt.show()
```

**Subplots**

* A subplot is a smaller plot (Axes) inside a larger Figure.

* Instead of drawing only one chart, you can organize multiple plots in rows and columns inside a single figure.

* Useful for comparing multiple variables or showing different aspects of the same dataset.

**Figure and Axes**

Figure → The overall canvas/window that contains plots.

Axes → The actual plots (subplots) inside the figure.

**Methods to Create Subplots**

* plt.subplot()

Syntax: plt.subplot(nrows, ncols, index)

nrows → number of rows in grid

ncols → number of columns

index → position of current subplot (starts at 1)

* plt.subplots()

More powerful and flexible.

Returns a Figure object and one/multiple Axes objects.

Syntax:  fig, axs = plt.subplots(nrows, ncols, figsize=(width, height))

fig → entire figure

axs → array (or single object) of Axes (subplots)


**MULTIPLE PLOTS**

* Sometimes, you want to compare multiple charts in one figure.

* Use subplot() function.

```python
x = [1, 2, 3, 4, 5]
y = [10, 20, 30, 40, 50]

plt.subplot(1, 2, 1)   # 1 row, 2 columns, 1st plot
plt.plot(x, y, color="red")
plt.title("Line Plot")

plt.subplot(1, 2, 2)   # 1 row, 2 columns, 2nd plot
plt.bar(x, y, color="blue")
plt.title("Bar Chart")



plt.tight_layout()
plt.show()
```

```python
# 6. Pie Chart — Sales Share by Cash Type

coffee_df["money"] = pd.to_numeric(coffee_df["money"], errors="coerce") #money datatype is int hence used errors=coerce

cash_sales = coffee_df.groupby("cash_type")["money"].sum()

plt.figure(figsize=(8,8))
plt.pie(
    cash_sales.values,
    labels=cash_sales.index,
    autopct="%1.1f%%", #to show the %age value
    startangle=90
)

plt.title("Sales share by cash type")
# plt.tight_layout()   #to avoid overlapping in multiple charts
plt.show()
```

```python
# 8. Scatter Plot — Hour vs Money
plt.figure(figsize=(10,6))
plt.scatter(
    coffee_df["hour_of_day"],
    coffee_df["money"],
    alpha=0.5
)

plt.title("Hour of the day vs Sales amount")
plt.xlabel("Hour of the day")
plt.ylabel("Sales amount")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# 9. Scatter Plot — Weekday vs Money
# Sort weekdays using Weekdaysort
weekday_order = (
    coffee_df[["Weekday","Weekdaysort"]].drop_duplicates().sort_values("Weekdaysort")
)

# Convert weekday names into numerical positions
weekday_mapping = dict(
    zip(
        weekday_order["Weekday"],
        range(len(weekday_order))
    )
)

weekday_numbers = coffee_df["Weekday"].map(weekday_mapping)

plt.figure(figsize=(10,6))

plt.scatter(
    weekday_numbers,
    coffee_df["money"],
    alpha=0.5
)

plt.title("Weekday vs Sales Amount")
plt.xlabel("Weekday")
plt.ylabel("Sales Amount")

plt.xticks(range(len(weekday_order)), weekday_order['Weekday'])

plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# 10.  Create a dashboard (using matplotlib subplots) with:
fig, axes = plt.subplots(
    2,
    2,
    figsize=(16,35)
)

# A bar chart of total revenue by coffee name.
coffee_revenue = (
    coffee_df.groupby("coffee_name")["money"].sum().sort_values(ascending=False)
)

axes[0,0].bar(
    coffee_revenue.index,
    coffee_revenue.values
)

axes[0,0].set_title("Total Revenue by coffee name")
axes[0,0].set_xlabel("Coffee name")
axes[0,0].set_ylabel("Total Revenue")

axes[0,0].tick_params(    # tick_params - the overall graph , whole graph
    axis="x",
    rotation=90
)

# A pie chart of sales share by cash type.
cash_sales = coffee_df.groupby("cash_type")["money"].sum()
axes[0,1].pie(
    cash_sales.values,
    labels=cash_sales.index,
    autopct="%1.1f%%",
    startangle=90
)

axes[0,1].set_title("Sales share by cash type")


# A histogram of sales by hour of the day.
axes[1,0].hist(
    coffee_df["hour_of_day"],
    bins=range(
        int(coffee_df["hour_of_day"].min()),
        int(coffee_df["hour_of_day"].max()) + 2
        ), 
        edgecolor="black"
    )

axes[1,0].set_title("Distribution of sales by hour")
axes[1,0].set_xlabel("Hour of day")
axes[1,0].set_ylabel("Number of transactions")

axes[1,0].set_xticks(
    range(
        int(coffee_df["hour_of_day"].min()),
        int(coffee_df["hour_of_day"].max()) + 1
    )
)

# A line plot of monthly sales trend.
monthly_sales = (
    coffee_df.groupby("Monthsort")["money"].sum().sort_index()
)

# Get month names in correct order
month_order = (
    coffee_df[["Monthsort","Month_name"]].drop_duplicates().sort_values("Monthsort")
)

axes[1,1].plot(
    monthly_sales.index,
    monthly_sales.values,
    marker = "o"
)

axes[1,1].set_title("Monthly sales trend")
axes[1,1].set_xlabel("Month")
axes[1,1].set_ylabel("Total revenue")

axes[1,1].set_title("Monthly sales trend")
axes[1,1].set_xlabel("Month")
axes[1,1].set_ylabel("Total revenue")

axes[1,1].set_xticks(monthly_sales.index)

axes[1,1].set_xticklabels(
    month_order["Month_name"],
    rotation=45
)

axes[1,1].grid(True)


# Dashboard title

fig.suptitle(
    "Coffee sales analysis dashboard",
    fontsize=20,
    fontweight="bold"
)

plt.tight_layout(
    rect=[0,0,1,0.96]
)

plt.show()
```

```python
# 1. Line Plot – Year-wise Total Sales Trend

superstore_df["Order Date"] = pd.to_datetime(superstore_df["Order Date"])

# Extract year
superstore_df["Year"] = superstore_df["Order Date"].dt.year.astype(int)

yearly_sales_trend = superstore_df.groupby("Year")["Sales"].sum().reset_index()
yearly_sales_trend["Year"] = yearly_sales_trend["Year"].astype(int)

print(yearly_sales_trend)

# plot
plt.figure(figsize=(10,6))

plt.plot(
    yearly_sales_trend["Year"],
    yearly_sales_trend["Sales"],
    marker="o"
)


plt.title("Yearwise total sales trend")
plt.xlabel("Year")
plt.ylabel("Total sales")

# plt.grid(True)
plt.xticks(yearly_sales_trend["Year"])
plt.tight_layout()
plt.show()
```

```python
# 3. Bar Chart – Top 10 Cities by Sales
cities_sales = (
    superstore_df.groupby("City")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print(cities_sales)

# plot
plt.figure(figsize=(10,8))

plt.barh(
    cities_sales.index,
    cities_sales.values
)

plt.title("Top 10 cities by sales")
plt.xlabel("City")
plt.ylabel("Total sales")

# plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# 4. Bar Chart – Top 10 States by Profit
profit_cities_sales = (
    superstore_df.groupby("City")["Profit"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print(profit_cities_sales)

# plot
plt.figure(figsize=(10,8))

plt.barh(
    profit_cities_sales.index,
    profit_cities_sales.values
)

plt.title("Top 10 cities by Profit")
plt.xlabel("City")
plt.ylabel("Total Profit")

# plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# 5. Pie Chart – Region-wise Sales Share
region_sales = superstore_df.groupby("Region")["Sales"].sum()

print(region_sales)

# pie chart
colors = ["#f07d7d", "#1663B0", "#dee794", "#ee8822"]

plt.figure(figsize=(10,8))

plt.pie(
    region_sales.values,
    labels=region_sales.index,
    autopct="%1.1f%%",
    startangle=90,
    colors=colors
)

plt.title("Region wise sales share")
plt.tight_layout()
plt.show()
```

```python
# 9. Line Plot – Monthly Sales Trend (All Years Combined)

# Convert Order Date to datetime
superstore_df["Order Date"] = pd.to_datetime(superstore_df["Order Date"])

# Extract month number
superstore_df["Month"] = superstore_df["Order Date"].dt.month

# Calculate total sales for each month across all years
monthly_sales = (
    superstore_df.groupby("Month")["Sales"]
      .sum()
      .sort_index()
)

# Month names
month_names = [
    "Jan", "Feb", "Mar", "Apr",
    "May", "Jun", "Jul", "Aug",
    "Sep", "Oct", "Nov", "Dec"
]

# Plot
plt.figure(figsize=(12, 6))

plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    marker="o"
)

plt.title("Monthly Sales Trend (All Years Combined)")
plt.xlabel("Month")
plt.ylabel("Total Sales")

plt.xticks(
    range(1, 13),
    month_names
)

plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# 10. Bar Chart – Loss-Making Sub-Categories

# Calculate total profit by sub-category
subcategory_profit = (
    superstore_df.groupby('Sub-Category')['Profit']
      .sum()
      .sort_values()
)

# Keep only loss-making sub-categories
loss_making = subcategory_profit[subcategory_profit < 0]

# Plot
plt.figure(figsize=(10, 6))
plt.bar(loss_making.index, loss_making.values)

plt.title('Loss-Making Sub-Categories')
plt.xlabel('Sub-Category')
plt.ylabel('Total Profit')
plt.xticks(rotation=45)
plt.axhline(0, linewidth=1)

plt.tight_layout()
plt.show()
```

```python
# Calculate total profit by sub-category
subcategory_profit = superstore_df.groupby('Sub-Category')['Profit'].sum()

# Keep only loss-making sub-categories
loss_making = subcategory_profit[subcategory_profit < 0]

# Convert losses to positive values for visualization
loss_values = loss_making.abs().sort_values(ascending=False)

# Plot
plt.figure(figsize=(10, 6))
bars = plt.bar(loss_values.index, loss_values.values)

plt.title('Loss-Making Sub-Categories')
plt.xlabel('Sub-Category')
plt.ylabel('Loss Amount')
plt.xticks(rotation=45)

# Add values on top of bars
for bar in bars:
    value = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        value,
        f'{value:,.0f}',
        ha='center',
        va='bottom'
    )

plt.tight_layout()
plt.show()
```

```python
# 11. 2 x 1 Dashboard
# Orders per Year – bar
# Profit per Year – line

# Orders per year — count unique orders
orders_per_year = (
    superstore_df.groupby('Year')['Order ID']
      .nunique()
)

# Profit per year — sum of profit
profit_per_year = (
    superstore_df.groupby('Year')['Profit']
      .sum()
)

# Create 2 x 1 dashboard
fig, axes = plt.subplots(2, 1, figsize=(10, 10))

# -------------------------
# 1. Orders per Year - Bar
# -------------------------
axes[0].bar(
    orders_per_year.index,
    orders_per_year.values
)

axes[0].set_title('Orders per Year')
axes[0].set_xlabel('Year')
axes[0].set_ylabel('Number of Orders')

# Add values above bars
for year, value in orders_per_year.items():
    axes[0].text(
        year,
        value,
        f'{value:,}',
        ha='center',
        va='bottom'
    )

# -------------------------
# 2. Profit per Year - Line
# -------------------------
axes[1].plot(
    profit_per_year.index,
    profit_per_year.values,
    marker='o'
)

axes[1].set_title('Profit per Year')
axes[1].set_xlabel('Year')
axes[1].set_ylabel('Total Profit')

# Add values to line points
for year, value in profit_per_year.items():
    axes[1].text(
        year,
        value,
        f'{value:,.0f}',
        ha='center',
        va='bottom'
    )

# Overall dashboard title
fig.suptitle(
    'Superstore Annual Performance Dashboard',
    fontsize=16,
    fontweight='bold'
)

plt.tight_layout()
plt.show()
```

```python
# 12. 2 x 2 Dashboard
# Sales by Category – bar
# Profit by Region – bar
# Monthly Sales Trend – line
# Segment-wise Sales Distribution – pie

# Create Year-Month for monthly trend
superstore_df['Year-Month'] = superstore_df['Order Date'].dt.to_period('M')

# -----------------------------
# 1. Sales by Category
# -----------------------------
sales_category = superstore_df.groupby('Category')['Sales'].sum()

# -----------------------------
# 2. Profit by Region
# -----------------------------
profit_region = superstore_df.groupby('Region')['Profit'].sum()

# -----------------------------
# 3. Monthly Sales Trend
# -----------------------------
monthly_sales = superstore_df.groupby('Year-Month')['Sales'].sum()

# -----------------------------
# 4. Segment-wise Sales
# -----------------------------
segment_sales = superstore_df.groupby('Segment')['Sales'].sum()

# -----------------------------
# Create 2 x 2 Dashboard
# -----------------------------
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# =================================
# Top-left: Sales by Category - Bar
# =================================
bars = axes[0, 0].bar(
    sales_category.index,
    sales_category.values
)

axes[0, 0].set_title('Sales by Category')
axes[0, 0].set_xlabel('Category')
axes[0, 0].set_ylabel('Sales')

for bar in bars:
    value = bar.get_height()
    axes[0, 0].text(
        bar.get_x() + bar.get_width() / 2,
        value,
        f'${value:,.0f}',
        ha='center',
        va='bottom'
    )

# =================================
# Top-right: Profit by Region - Bar
# =================================
bars = axes[0, 1].bar(
    profit_region.index,
    profit_region.values
)

axes[0, 1].set_title('Profit by Region')
axes[0, 1].set_xlabel('Region')
axes[0, 1].set_ylabel('Profit')

# Zero reference line
axes[0, 1].axhline(0, linewidth=1)

for bar in bars:
    value = bar.get_height()

    # Put label above positive bars and below negative bars
    if value >= 0:
        va = 'bottom'
        offset = 100
    else:
        va = 'top'
        offset = -100

    axes[0, 1].text(
        bar.get_x() + bar.get_width() / 2,
        value + offset,
        f'${value:,.0f}',
        ha='center',
        va=va
    )

# =================================
# Bottom-left: Monthly Sales - Line
# =================================
axes[1, 0].plot(
    monthly_sales.index.astype(str),
    monthly_sales.values,
    marker='o'
)

axes[1, 0].set_title('Monthly Sales Trend')
axes[1, 0].set_xlabel('Month')
axes[1, 0].set_ylabel('Sales')

# Show fewer x-axis labels for readability
step = max(1, len(monthly_sales) // 12)
axes[1, 0].set_xticks(
    range(0, len(monthly_sales), step)
)
axes[1, 0].set_xticklabels(
    monthly_sales.index.astype(str)[::step],
    rotation=45,
    ha='right'
)

# =================================
# Bottom-right: Segment Sales - Pie
# =================================
axes[1, 1].pie(
    segment_sales.values,
    labels=segment_sales.index,
    autopct='%1.1f%%',
    startangle=90
)

axes[1, 1].set_title('Segment-wise Sales Distribution')

# =================================
# Overall Dashboard
# =================================
fig.suptitle(
    'Superstore Sales & Profit Dashboard',
    fontsize=18,
    fontweight='bold'
)

plt.tight_layout()
plt.show()
```

```python
# 13. Create a 2×2 Dashboard:
# Top 5 Sub-Categories by Sales – bar
# Region-wise Profit Share – pie
# Yearly Sales Trend – line
# Discount vs Profit – scatter

# -----------------------------------
# 1. Top 5 Sub-Categories by Sales
# -----------------------------------
top_5_subcategories = (
    superstore_df.groupby('Sub-Category')['Sales']
      .sum()
      .sort_values(ascending=False)
      .head(5)
)

# -----------------------------------
# 2. Region-wise Profit Share
# -----------------------------------
region_profit = (
    superstore_df.groupby('Region')['Profit']
      .sum()
)

# -----------------------------------
# 3. Yearly Sales Trend
# -----------------------------------
yearly_sales = (
    superstore_df.groupby('Year')['Sales']
      .sum()
      .sort_index()
)

# -----------------------------------
# 4. Discount vs Profit
# -----------------------------------
discount_profit = superstore_df[['Discount', 'Profit']].dropna()

# -----------------------------------
# Create 2 × 2 Dashboard
# -----------------------------------
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# ===================================
# Top-left: Top 5 Sub-Categories
# ===================================
bars = axes[0, 0].bar(
    top_5_subcategories.index,
    top_5_subcategories.values
)

axes[0, 0].set_title('Top 5 Sub-Categories by Sales')
axes[0, 0].set_xlabel('Sub-Category')
axes[0, 0].set_ylabel('Total Sales')
axes[0, 0].tick_params(axis='x', rotation=45)

# Add values
for bar in bars:
    value = bar.get_height()
    axes[0, 0].text(
        bar.get_x() + bar.get_width() / 2,
        value,
        f'${value:,.0f}',
        ha='center',
        va='bottom'
    )


# ===================================
# Top-right: Region-wise Profit Share
# ===================================
axes[0, 1].pie(
    region_profit.values,
    labels=region_profit.index,
    autopct='%1.1f%%',
    startangle=90
)

axes[0, 1].set_title('Region-wise Profit Share')


# ===================================
# Bottom-left: Yearly Sales Trend
# ===================================
axes[1, 0].plot(
    yearly_sales.index,
    yearly_sales.values,
    marker='o'
)

axes[1, 0].set_title('Yearly Sales Trend')
axes[1, 0].set_xlabel('Year')
axes[1, 0].set_ylabel('Total Sales')
axes[1, 0].set_xticks(yearly_sales.index)

# Add values to points
for year, value in yearly_sales.items():
    axes[1, 0].text(
        year,
        value,
        f'${value:,.0f}',
        ha='center',
        va='bottom'
    )

# ===================================
# Bottom-right: Discount vs Profit
# ===================================
axes[1, 1].scatter(
    discount_profit['Discount'],
    discount_profit['Profit'],
    alpha=0.4
)

axes[1, 1].set_title('Discount vs Profit')
axes[1, 1].set_xlabel('Discount')
axes[1, 1].set_ylabel('Profit')

# Reference lines
axes[1, 1].axhline(0, linewidth=1)
axes[1, 1].axvline(0, linewidth=1)


# ===================================
# Overall Dashboard
# ===================================
fig.suptitle(
    'Superstore Sales & Profit Dashboard',
    fontsize=18,
    fontweight='bold'
)

plt.tight_layout()
plt.show()
```
