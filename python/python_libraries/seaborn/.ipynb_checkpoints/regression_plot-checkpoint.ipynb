**REGRESSION PLOT** - show data points and the best-fit line that represents the relationship between variables.

REGRESSION  

 statistical method used to understand the relationship between variables. predict the value of dependent variable based on the values of independent variables.

* **regplot**() -> Scatter plot with regression line


syntax

sns.regplot(

    data=None,       # Your DataFrame
    x=None,          # Column name for x-axis (numeric)
    y=None,          # Column name for y-axis (numeric)
    ci=95,           # Confidence interval for regression line (default 95%)
    scatter=True,    # Show scatter points (True/False)
    fit_reg=True,    # Show regression line (True/False)
    order=1,         # Polynomial order (1=linear, 2=quadratic, etc.)
    logx=False,      # Apply log scale on x-axis (True/False)
    color=None       # Color of points and line
)

**log scale**

* The axis increases by powers of 10 (1, 10, 100, 1000,
* This means small values are stretched apart and large values are compressed.

**Why Use Log Scale?**

* When your data spans a very large range (e.g., from 1 to 10000).

* To make patterns easier to see in skewed data.

* To handle exponential relationships between variables.

**polynomial** - Type of regression , modeled as a polynomial (curved) function rather than a straight line.

**Why Use Polynomial Regression?**

* When the relationship between variables is not linear.
* Captures curved trends in data.

 **Confidence Interval (CI)**

A confidence interval is a range of values that is likely to contain the true value of a parameter (like a regression line) with a certain level of confidence (usually 95%).

* If you collect the same data many times and fit regression lines, most of those lines would fall within the shaded confidence interval.

* For a 95% CI, we are 95% confident the true regression line lies inside that shaded region.

**Why Important?**

* Shows uncertainty around the regression line.

* Wide CI → predictions are less reliable.

* Narrow CI → predictions are more reliable.


 **lmplot()**: Combines regression and faceting for multiple subsets

syntax:

sns.lmplot(

    data=None,       # Your DataFrame
    x=None,          # Column for x-axis
    y=None,          # Column for y-axis
    hue=None,        # Categorical variable to color by
    col=None,        # Column to facet by
    row=None,        # Row to facet by
    ci=95,           # Confidence interval
    scatter=True,    # Show scatter points
    order=1,         # Polynomial regression order
    logx=False,      # Log scale for x-axis
    palette=None     # Color palette for hue
)




```python
# Basic Regression Plot
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
tips = sns.load_dataset("tips")

sns.regplot(x="total_bill", y="tip", data=tips)

plt.title("Basic Regression Plot")
plt.show()
```

```python
# Regression Line Only (No Scatter)
sns.regplot(x="total_bill", y="tip", data=tips, scatter=False, color="red")
plt.title("Regression Line Only")
plt.show()
```

```python
# Polynomial Regression (Order 2)
sns.regplot(x="total_bill", y="tip", data=tips, order=2, line_kws={"color":"green"})
plt.title("Polynomial Regression (Quadratic Fit)")
plt.show()
```

```python
# Confidence Interval Disabled
sns.regplot(x="total_bill", y="tip", data=tips, ci=95)

plt.title("Regression Plot without Confidence Interval")
plt.show()
```

```python
# Multiple Categories using lmplot
sns.lmplot(x="total_bill", y="tip", hue="smoker", data=tips, height=6, aspect=1.3)
plt.title("Regression Plot Grouped by Smoker/Non-Smoker")
plt.show()
```

```python
sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    col="day"
)
```

```python
sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    row="sex"
)
```

```python
sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    row="sex",
    col="time"
)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
car_df = pd.read_csv("CarPrice_Assignment.csv")
```

```python
# 1. Plot a regression plot between enginesize and price.
sns.regplot(data=car_df,x='enginesize', y='price')

plt.title('Engine Size vs Car Price')
plt.xlabel('Engine Size')
plt.ylabel('Price')
plt.show()
```

```python
# 2. Plot a regression plot between horsepower and price with a 95% confidence interval.

sns.regplot(
    data=car_df,
    x="horsepower",
    y="price",
    ci=95)

plt.title('Horsepower vs Car Price')
plt.xlabel('Horsepower')
plt.ylabel('Price')
plt.show()
```

```python
# 3. Plot regression between citympg and price using a log scale for the x-axis.
sns.regplot(data=car_df, x='citympg', y='price')

plt.xscale('log')
plt.title('City MPG vs Car Price')
plt.xlabel('City MPG (log scale)')
plt.ylabel('Price')
plt.show()
```

```python
# 4. Plot regression between enginesize and price using a polynomial fit of degree 2.
sns.regplot(data=car_df, x='enginesize', y='price', order=2)

plt.title('Engine Size vs Car Price (Polynomial Regression)')
plt.xlabel('Engine Size')
plt.ylabel('Price')
plt.show()
```

```python
# 5. Plot regression between horsepower and price, colored by carbody.
sns.lmplot(
    data=car_df,
    x='horsepower',
    y='price',
    hue='carbody',
    height=6,
    aspect=1.2
)

plt.title('Horsepower vs Car Price by Car Body')
plt.xlabel('Horsepower')
plt.ylabel('Price')
plt.show()
```

```python
# 6. Plot Wheelbase vs Price with regression line only (no scatter points).
sns.regplot(
    data=car_df,
    x='wheelbase',
    y='price',
    scatter=False
)

plt.title('Wheelbase vs Car Price')
plt.xlabel('Wheelbase')
plt.ylabel('Price')
plt.show()
```

```python
# 7. Plot Highway MPG vs Price and remove confidence interval.
sns.regplot(
    data=car_df,
    x='highwaympg',
    y='price',
    ci=None
)

plt.title('Highway MPG vs Car Price')
plt.xlabel('Highway MPG')
plt.ylabel('Price')
plt.show()
```

```python
# 8. Plot Horsepower vs Price by Fuel Type.
sns.lmplot(
    data=car_df,
    x='horsepower',
    y='price',
    hue='fueltype',
    height=6,
    aspect=1.2
)

plt.title('Horsepower vs Car Price by Fuel Type')
plt.xlabel('Horsepower')
plt.ylabel('Price')
plt.show()
```

```python
# 9. Plot Peak RPM vs Price with custom color (red).
sns.regplot(
    data=car_df,
    x='peakrpm',
    y='price',
    color='red'
)

plt.title('Peak RPM vs Car Price')
plt.xlabel('Peak RPM')
plt.ylabel('Price')
plt.show()
```

```python
# 10. Plot Engine Size vs Price with Facets by Aspiration.
sns.lmplot(
    data=car_df,
    x='enginesize',
    y='price',
    col='aspiration',
    height=5,
    aspect=1.1
)

plt.show()
```
