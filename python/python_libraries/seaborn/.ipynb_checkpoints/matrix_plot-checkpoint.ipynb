5. **MATRIX PLOT** - Two-dimensional representation of data in the form of a grid (rows × columns), where the values inside the grid are represented using colors or annotations.Used for correlation and heatmaps



* **Heatmap**(): A heatmap uses color intensity to represent values in a 2D table.used to visualize correlations, frequencies, or pivot tables.Correlation heatmap for numeric columns in a dataset.

when to use :  When you want to visualize correlation or compare values in a matrix form.

syntax:

sns.heatmap(data, annot=True, cmap="coolwarm")

data → 2D data (like correlation matrix or pivot table).

annot=True → shows the values inside the cells.

cmap → color scheme (e.g., "coolwarm", "viridis", "Blues").

* **Cluster Map ()**: Similar to a heatmap, but also applies hierarchical clustering to group similar rows/columns together.Helps in pattern detection.
find the group with similar features

when to use : When you want to identify patterns and natural groupings in your data.

syntax

sns.clustermap(data, cmap="mako", standard_scale=1)

cmap → color map.

standard_scale → normalizes rows/columns.

**correlation** - Correlation is a statistical measure that tells us how two variables are related to each other.

**Types of Correlation**

* Positive Correlation - When one variable increases, the other also increases.
* Negative Correlation - When one variable increases, the other decreases.
* No Correlation - Variables do not show any clear relationship.

**Why Correlation is Useful**

* Helps to find relationships between variables before building models.
* Helps identify which variables move together or are independent.
* Used in heatmaps, feature selection, and clustering.

**Correlation Coefficient**

A correlation coefficient is a numerical value that measures the strength and direction of a relationship between two variables.

Denoted by r, ranges from -1 to 1.

r value	Meaning
1	Perfect positive correlation
-1	Perfect negative correlation
0	No correlation

To calculate correlation :

corr = dataframe[column].corr()




**Color Map Name	Description / Usage**
* 'coolwarm'	  Blue = negative, Red = positive
* 'viridis'	    Green → Yellow, good for sequential data
* 'plasma'	    Dark → Bright, sequential
* 'YlGnBu'    	Yellow → Green → Blue, sequential
* 'Reds'	      Shades of red, good for intensity
* 'Blues'       	Shades of blue, good for correlations
* 'RdBu_r'     	Red ↔ Blue diverging, centered around 0
* 'magma'	    Dark → Light, sequential

```python
# Heatmap
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# Create random matrix
data = np.random.rand(6, 6)

# Heatmap
sns.heatmap(data, annot=True, cmap="YlGnBu")

plt.title("Heatmap Example")
plt.show()

# annot=True writes numbers on the heatmap.
# cmap controls color (examples: "coolwarm", "viridis", "YlGnBu").
```

```python
# Correlation Heatmap (Real Dataset)
# Load dataset
tips = sns.load_dataset("tips")

# Select only numerical columns
numeric_tips = tips.select_dtypes(include=['number'])

# Correlation matrix
corr = numeric_tips.corr()

# Heatmap of correlation
sns.heatmap(corr, annot=True, cmap="coolwarm", linewidths=0.5)

plt.title("Correlation Heatmap of Tips Dataset")
plt.show()
```

```python
# Masked Heatmap (Half Matrix)
mask = np.triu(np.ones_like(corr, dtype=bool))  # Upper triangle mask

sns.heatmap(corr, annot=True, cmap="coolwarm", mask=mask)

plt.title("Masked Heatmap (Lower Triangle Only)")
plt.show()

# Helps when correlation matrix is symmetric.
```

```python
# Clustermap
sns.clustermap(corr, annot=True, cmap="viridis")

plt.title("Cluster Map of Correlations")
plt.show()

# Automatically groups similar rows and columns together.
```

```python
# convert 'dt' column to datetime
temp_df["dt"] = pd.to_datetime(temp_df['dt'], errors="coerce")

# extract year and month
temp_df["Year"] = temp_df["dt"].dt.year
temp_df["Month"] = temp_df["dt"].dt.month

# set seaborn style
sns.set_style("whitegrid")
```

```python
# 1. Show Year vs Month heatmap of Land Average Temperature

# Get the most recent year in the dataset
latest_year = temp_df["Year"].max()

print(latest_year)
# Keep only the last 50 years
recent_df = temp_df[
    temp_df["Year"] >= latest_year - 5
]

# Create Year x Month matrix
land_avg = recent_df.pivot_table(
    index="Year",
    columns="Month",
    values="LandAverageTemperature"
)

# Plot heatmap
plt.figure(figsize=(7, 5))

ax = sns.heatmap(
    land_avg,
    cmap="coolwarm",
    annot=True,              # Print values inside cells
    fmt=".1f",               # Show 1 decimal place
    linewidths=0.3,
    cbar_kws={"label": "Temperature (°C)"}
)

# Month names
import calendar

ax.set_xticklabels(
    [calendar.month_abbr[i] for i in range(1, 13)]
)

plt.title("Year vs Month - Land Average Temperature (Last 10 Years)")
plt.xlabel("Month")
plt.ylabel("Year")

plt.tight_layout()
plt.show()
```

```python
# 2. Plot Year vs Month heatmap of Land Max Temperature
# Create Year x Month matrix

heatmap_data = recent_df.pivot_table(
    index="Year",
    columns="Month",
    values="LandMaxTemperature"
)

# Create heatmap
plt.figure(figsize=(7, 5))

ax = sns.heatmap(
    heatmap_data,
    cmap="YlOrRd",
    annot=True,                 # Print temperature values
    fmt=".1f",                  # Show 1 decimal place
    linewidths=0.3,
    cbar_kws={"label": "Temperature (°C)"}
)

# Change month numbers to month names
ax.set_xticklabels(
    [calendar.month_abbr[i] for i in range(1, 13)]
)

# Add title and labels
plt.title(
    "Year vs Month - Land Max Temperature (Recent 20 Years)",
    fontsize=14
)

plt.xlabel("Month")
plt.ylabel("Year")

plt.tight_layout()
plt.show()
```

```python
# 3. Plot Year vs Month heatmap of Land Min Temperature
heatmap_data = recent_df.pivot_table(
    index='Year',
    columns='Month',
    values='LandMinTemperature'
)

plt.figure(figsize=(7, 5))
ax = sns.heatmap(
    heatmap_data,
    cmap="YlOrRd",
    annot=True,                 # Print temperature values
    fmt=".1f",                  # Show 1 decimal place
    linewidths=0.3,
    cbar_kws={"label": "Temperature (°C)"}
)

# Change month numbers to month names
ax.set_xticklabels(
    [calendar.month_abbr[i] for i in range(1, 13)]
)

# Add title and labels
plt.title(
    "Year vs Month - Land Min Temperature",
    fontsize=14
)
plt.xlabel("Month")
plt.ylabel("Year")
plt.tight_layout()
plt.show()
```

```python
# 5. Heatmap of Land Average Temperature Uncertainty
import calendar

# Get the latest year in the dataset
latest_year = temp_df["Year"].max()

# Select the most recent 20 years
recent_df = temp_df[
    temp_df["Year"] >= latest_year - 10
]

# Create Year x Month matrix
heatmap_data = recent_df.pivot_table(
    index="Year",
    columns="Month",
    values="LandAverageTemperatureUncertainty"
)

# Create heatmap
plt.figure(figsize=(7, 5))

ax = sns.heatmap(
    heatmap_data,
    cmap="viridis",
    annot=True,                 # Print uncertainty values
    fmt=".2f",                  # Show 2 decimal places
    linewidths=0.3,
    cbar_kws={"label": "Temperature Uncertainty (°C)"}
)

# Change month numbers to month names
ax.set_xticklabels(
    [calendar.month_abbr[i] for i in range(1, 13)]
)

# Title and labels
plt.title(
    "Year vs Month - Land Average Temperature Uncertainty (Recent 10 Years)",
    fontsize=14
)

plt.xlabel("Month")
plt.ylabel("Year")

plt.tight_layout()
plt.show()
```

```python
# 6. Compare all temperature measures (Year vs Variable)

# Select temperature columns
temperature_columns = [
    "LandAverageTemperature",
    "LandMaxTemperature",
    "LandMinTemperature",
    "LandAndOceanAverageTemperature"
]

# Get the latest year
latest_year = temp_df["Year"].max()

# Select the most recent 10 years
recent_df = temp_df[
    temp_df["Year"] >= latest_year - 10
]

# Calculate yearly average temperature
yearly_temperature = recent_df.groupby("Year")[temperature_columns].mean()

# Rename columns for easier reading
yearly_temperature.columns = [
    "Land Average",
    "Land Maximum",
    "Land Minimum",
    "Land & Ocean Average"
]

# Create clustered heatmap
g = sns.clustermap(
    yearly_temperature,
    cmap="coolwarm",
    figsize=(7, 5),
    annot=True,              # Show temperature values
    fmt=".1f",               # One decimal place
    linewidths=0.3,
    cbar_kws={"label": "Temperature (°C)"},
    row_cluster=False        # Keep years in chronological order
)

# Add title
g.fig.suptitle(
    "Year vs Temperature Variables - Recent 10 Years",
    fontsize=14,
    y=1.02
)

plt.show()
```

```python
# 7. Compare multiple temperature measures (Year vs Variable)
# Select temperature columns
temperature_columns = [
    "LandAverageTemperature",
    "LandMaxTemperature",
    "LandMinTemperature",
    "LandAndOceanAverageTemperature"
]

# Calculate yearly average temperature
yearly_temperature = recent_df.groupby("Year")[temperature_columns].mean()

# Rename columns for easier reading
yearly_temperature.columns = [
    "Land Average",
    "Land Maximum",
    "Land Minimum",
    "Land & Ocean Average"
]

# Transpose the matrix
# Rows = temperature variables
# Columns = years
temperature_matrix = yearly_temperature.T

# Create clustered heatmap
g = sns.clustermap(
    temperature_matrix,
    cmap="coolwarm",
    figsize=(7, 5),
    annot=True,                 # Show actual temperature values
    fmt=".1f",                  # One decimal place
    linewidths=0.3,
    cbar_kws={"label": "Temperature (°C)"},
    col_cluster=False           # Keep years in chronological order
)

# Add title
g.fig.suptitle(
    "Temperature Variables vs Year - Recent 10 Years",
    fontsize=14,
    y=1.02
)
plt.show()
```

```python
# 8. Plot Year vs Month heatmap of Land & Ocean Avg Temperature
# This is the same matrix as #4,
# included separately because it was requested.
# Create Year x Month matrix
land_ocean_matrix = recent_df.pivot_table(
    index="Year",
    columns="Month",
    values="LandAndOceanAverageTemperature"
)

# Create heatmap
plt.figure(figsize=(7, 5))

ax = sns.heatmap(
    land_ocean_matrix,
    cmap="coolwarm",
    annot=True,                 # Print temperature values
    fmt=".1f",                  # Show 1 decimal place
    linewidths=0.3,
    cbar_kws={"label": "Temperature (°C)"}
)

# Change month numbers to month names
ax.set_xticklabels(
    [calendar.month_abbr[i] for i in range(1, 13)]
)

# Title and labels
plt.title(
    "Year vs Month - Land & Ocean Average Temperature (Recent 10 Years)",
    fontsize=14
)

plt.xlabel("Month")
plt.ylabel("Year")
plt.tight_layout()
plt.show()
```

**DATA MERGING AND JOINING**

* In real-world datasets, information may be spread across multiple tables (DataFrames).

* To analyze properly, we need to combine them into one DataFrame.

* Pandas provides tools for this:

* Method	Use Case

merge()	Combine two DataFrames on a common column (like SQL joins).

concat()	Stack two DataFrames vertically or horizontally.

join()	Similar to merge, but works with index instead of columns.

* Once merged, we can easily use Seaborn for visualization.

**Types of Merge (like SQL joins)**

Inner Join → Keep only matching rows.

Left Join → Keep all rows from the left DataFrame.

Right Join → Keep all rows from the right DataFrame.

Outer Join → Keep all rows from both DataFrames.

image.png

```python
# Merge average tip by day with average total bill

import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

avg_tip = tips.groupby("day")["tip"].mean().reset_index()
avg_bill = tips.groupby("day")["total_bill"].mean().reset_index()

merged = pd.merge(avg_tip, avg_bill, on="day")

sns.barplot(data=merged, x="day", y="tip", color="blue", label="Avg Tip")
sns.barplot(data=merged, x="day", y="total_bill", color="orange", alpha=0.5, label="Avg Bill")
plt.legend()
plt.title("Average Tip and Total Bill by Day")
plt.show()

# Compares average tips vs bills for each day.
```
