**CATEGORICAL PLOT**

* it deals with categorical variables
* used for visualizing the relationship between variables.
* Those variables can be either be completely numerical or a category like a group, class or division.

**Functions for Categorical Plots in Seaborn**

1. **boxplot()** - Shows distribution (median, quartiles, outliers).the dots indicating the presence of outliers.

syntax :
sns.boxplot([x, y, hue, data, order, hue_order, …])


2. **violinplot()** - Combines boxplot + KDE ( kernal density estimation).estimate the probability distribution of a dataset (continuous variable).

syntax:
sns.violinplot([x, y, hue, data, order, …])

3. **barplot()** - shows mean value of numerical variable for each category (with error bars).

syntax :
sns.barplot([x, y, hue, data, order, hue_order, …])

4. **countplot()**- counts the categories and returns a count of their occurrences. Shows frequency of each category.

syntax:
sns.countplot([x, y, hue, data, order, …])

5. **Stripplo**t -creates a scatter plot based on  a category

syntax:
sns.stripplot([x, y, hue, data, order, …])

6. **swarmplot()** - Like stripplot, but adjusts points to avoid overlap.

syntax:
sns.swarmplot([x, y, hue, data, order, …])

7. **catplot()** - General function for categorical plots.kind can be: "strip", "swarm", "box", "violin", "boxen", "point", "bar", "count".

syntax:
sns.catplot([x, y , data , kind,...])

8. **pointplot()** - Shows mean value with lines connecting categories (good for trends).

syntax :
sns.pointplot([ x, y. data,..])

**KERNAL DENSITY ESTIMATE (KDE)**

 KDE is a way to estimate the probability density function (PDF) of a continuous variable.

 It can be used for both univariate and bivariate distributions, making it a versatile option for exploring data.

 * Univariate KDE: Visualizes the distribution of a single variable.

* Bivariate KDE: Displays the joint distribution of two variables.

**DISTRIBUTION PLOT**

* It provides a high-level interface for drawing attractive and informative statistical graphics.
* used for examining univariate and bivariate distributions.
* A distribution plot shows how data values are spread (their frequency, density, or probability) across a range.

**Common Types of Distribution Plots**

* **Histogram**  → Bars showing frequency counts.

syntax

sns.histplot(data=df, x="numeric_column", bins=20, kde=True)

* **Kernel Density Estimate (KDE)**  → Smooth curve estimate of probability density.

syntax

sns.kdeplot(data=df, x="numeric_column", hue="category_column", fill=True)


* **Combination (Hist + KDE)** → Both histogram & KDE together.

DATA PREPROCESSING : E-COMMERCE DATASET

```python
# load the dataset

# read csv file
ecomm_df = pd.read_csv("data.csv", encoding="latin-1")

# display basic information
print("Original shape - ",ecomm_df.shape)
```

```python
# ============================================================
# 3. Remove duplicate rows
# ============================================================

# Count duplicate rows
print("\nNumber of duplicate rows:", ecomm_df.duplicated().sum())

# Remove exact duplicate rows
ecomm_df = ecomm_df.drop_duplicates()

print("Shape after removing duplicates:", ecomm_df.shape)
```

```python
# ============================================================
# 4. Convert InvoiceDate to datetime
# ============================================================

# Convert InvoiceDate from object/string to datetime
ecomm_df["InvoiceDate"] = pd.to_datetime(
    ecomm_df["InvoiceDate"],
    errors="coerce"
)

# Check if any dates could not be converted
print("\nMissing/invalid InvoiceDate values:",
      ecomm_df["InvoiceDate"].isna().sum())
```

```python
# ============================================================
# 5. Handle missing Description values
# ============================================================

# Check missing descriptions
print("\nMissing Description values:",
      ecomm_df["Description"].isna().sum())

# Replace missing descriptions with "Unknown"
ecomm_df["Description"] = ecomm_df["Description"].fillna("Unknown")
```

```python
# ============================================================
# 6. Handle missing CustomerID
# ============================================================

# CustomerID is missing for many transactions.
# We should NOT blindly replace it with 0 because 0 would
# represent a fake customer.

print("\nMissing CustomerID values:",
      ecomm_df["CustomerID"].isna().sum())

# Create a separate label for missing customer IDs
ecomm_df["CustomerID"] = ecomm_df["CustomerID"].astype("Int64")

# Keep missing CustomerID as <NA>.
# For customer-level analysis, we can later remove these rows.
```

```python
# ============================================================
# 7. Check negative Quantity values
# ============================================================

# Count negative quantities
negative_quantity = (ecomm_df["Quantity"] < 0).sum()

print("\nNegative Quantity values:", negative_quantity)

# Check invoices beginning with "C"
# These generally represent cancelled/returned transactions.
cancelled_invoices = ecomm_df["InvoiceNo"].astype(str).str.startswith("C")

print("Cancellation invoices:", cancelled_invoices.sum())

# Check how many negative quantities are cancellation invoices
print(
    "Negative quantities in cancellation invoices:",
    ((ecomm_df["Quantity"] < 0) & cancelled_invoices).sum()
)
```

```python
# ============================================================
# 8. Handle negative Quantity values
# ============================================================

# For SALES analysis, remove returns/cancellations.
# We also remove any other negative quantities.

df_sales = ecomm_df[ecomm_df["Quantity"] > 0].copy()

print("\nShape of sales-only dataset:", df_sales.shape)

# IMPORTANT:
# Keep the original df if you want to analyze returns.
# Use df_sales for normal sales/revenue analysis.
```

```python
# ============================================================
# 9. Check negative UnitPrice values
# ============================================================

print("\nNegative UnitPrice values:")
print(ecomm_df[ecomm_df["UnitPrice"] < 0][
    ["InvoiceNo", "StockCode", "Description", "Quantity", "UnitPrice"]
])
```

```python
# ============================================================
# 10. Handle negative UnitPrice values
# ============================================================

# Negative prices are not valid for normal sales analysis.
# Remove them from the sales dataset.

df_sales = df_sales[df_sales["UnitPrice"] > 0].copy()

print(
    "\nShape after removing non-positive UnitPrice:",
    df_sales.shape
)
```

```python
# ============================================================
# 11. Handle zero UnitPrice
# ============================================================

# Count zero-price transactions
print(
    "\nZero UnitPrice values:",
    (df_sales["UnitPrice"] == 0).sum()
)

# Remove zero-price transactions for revenue analysis
df_sales = df_sales[df_sales["UnitPrice"] > 0].copy()
```

```python
# ============================================================
# 12. Check StockCode
# ============================================================

# Display most common StockCodes
print("\nTop StockCodes:")
print(df_sales["StockCode"].value_counts().head(10))

# Convert StockCode to string
df_sales["StockCode"] = df_sales["StockCode"].astype(str)
```

```python
# ============================================================
# 13. Clean Description text
# ============================================================

# Remove leading/trailing spaces
df_sales["Description"] = (
    df_sales["Description"]
    .astype(str)
    .str.strip()
)

# Convert descriptions to uppercase
df_sales["Description"] = (
    df_sales["Description"]
    .str.upper()
)
```

```python
# ============================================================
# 14. Clean Country column
# ============================================================

# Remove leading/trailing spaces
df_sales["Country"] = (
    df_sales["Country"]
    .astype(str)
    .str.strip()
)
```

```python
# ============================================================
# 15. Calculate Total Sales
# ============================================================

# Total revenue for each transaction
df_sales["TotalSales"] = (
    df_sales["Quantity"] *
    df_sales["UnitPrice"]
)

print("\nFirst 5 TotalSales values:")
print(df_sales[["Quantity", "UnitPrice", "TotalSales"]].head())
```

```python
# ============================================================
# 16. Create useful date columns
# ============================================================

# Extract date
df_sales["Date"] = df_sales["InvoiceDate"].dt.date

# Extract year
df_sales["Year"] = df_sales["InvoiceDate"].dt.year

# Extract month
df_sales["Month"] = df_sales["InvoiceDate"].dt.month

# Extract month name
df_sales["MonthName"] = df_sales["InvoiceDate"].dt.month_name()

# Extract day
df_sales["Day"] = df_sales["InvoiceDate"].dt.day

# Extract day of week
df_sales["DayOfWeek"] = df_sales["InvoiceDate"].dt.day_name()

# Extract hour
df_sales["Hour"] = df_sales["InvoiceDate"].dt.hour
```

```python
# ============================================================
# 17. Create a customer-analysis dataset
# ============================================================

# CustomerID is required for customer-level analysis.
# Therefore, remove transactions without CustomerID.

df_customer = df_sales.dropna(
    subset=["CustomerID"]
).copy()

print("\nShape for customer analysis:", df_customer.shape)
```

```python
# ============================================================
# 18. Convert CustomerID to integer
# ============================================================

df_customer["CustomerID"] = (
    df_customer["CustomerID"]
    .astype(int)
)
```

```python
# Set the Seaborn plotting style
sns.set_style("whitegrid")

# ============================================================
# LOAD CLEANED DATASET
# ============================================================

# Load the cleaned sales dataset
df_sales = pd.read_csv("ecommerce_sales_cleaned.csv")

# Convert InvoiceDate to datetime format
df_sales["InvoiceDate"] = pd.to_datetime(
    df_sales["InvoiceDate"],
    errors="coerce"
)

# Calculate total sales for each transaction
df_sales["TotalSales"] = (
    df_sales["Quantity"] * df_sales["UnitPrice"]
)

# Display basic information
print(df_sales.head())
print(df_sales.shape)
```

```python
# ============================================================
# 1. HISTOGRAM + KDE FOR UnitPrice
# ============================================================

plt.figure(figsize=(10, 6))

# Keep positive prices
positive_prices = df_sales[df_sales["UnitPrice"] > 0].copy()

# Plot only prices up to 100
price_filtered = positive_prices[positive_prices["UnitPrice"] <= 100]

plt.figure(figsize=(10, 6))

sns.histplot(
    data=price_filtered,
    x="UnitPrice",
    kde=True,
    bins=50
)

plt.title("Distribution of Unit Price (0–100)")
plt.xlabel("Unit Price")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 2. HISTOGRAM + KDE FOR Quantity
# ============================================================
plt.figure(figsize=(10, 6))

# Keep positive prices
positive_quantity = df_sales[df_sales["Quantity"] > 0].copy()

# Plot only prices up to 100
quantity_filtered = positive_quantity[positive_quantity["Quantity"] <= 100]

# Create histogram with KDE
sns.histplot(
    data=quantity_filtered,
    x="Quantity",
    bins=50,
    kde=True
)

plt.title("Distribution of Quantity")
plt.xlabel("Quantity")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 3. TOTAL SPEND PER CUSTOMER
#    Histogram + KDE
# ============================================================

# Remove transactions without CustomerID
customer_data = df_sales.dropna(
    subset=["CustomerID"]
).copy()

# Calculate total spending for each customer
customer_spend = (
    customer_data
    .groupby("CustomerID")["TotalSales"]
    .sum()
    .reset_index()
)

# Plot the distribution of customer spending
plt.figure(figsize=(10, 6))

sns.histplot(
    data=customer_spend,
    x="TotalSales",
    bins=50,
    kde=True
)

plt.title("Distribution of Total Spend per Customer")
plt.xlabel("Total Spend")
plt.ylabel("Number of Customers")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 4. TOTAL SALES BY COUNTRY
#    BAR CHART
# ============================================================

# Group sales by country
country_sales = (
    df_sales
    .groupby("Country")["TotalSales"]
    .sum()
    .sort_values(ascending=False)
)

# Plot total sales by country
plt.figure(figsize=(14, 8))

sns.barplot(
    x=country_sales.values,
    y=country_sales.index
)

plt.title("Total Sales by Country")
plt.xlabel("Total Sales")
plt.ylabel("Country")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 5. NUMBER OF ITEMS PER INVOICE
#    HISTOGRAM + KDE
# ============================================================

# Calculate total number of items in each invoice
items_per_invoice = (
    df_sales
    .groupby("InvoiceNo")["Quantity"]
    .sum()
    .reset_index(name="NumberOfItems")
)

# Plot distribution
plt.figure(figsize=(10, 6))

sns.histplot(
    data=items_per_invoice,
    x="NumberOfItems",
    bins=50,
    kde=True
)

plt.title("Distribution of Number of Items per Invoice")
plt.xlabel("Number of Items")
plt.ylabel("Number of Invoices")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 6. SALES TRENDS OVER TIME
#    AGGREGATE SALES BY DATE
# ============================================================

# Create a Date column
df_sales["Date"] = df_sales["InvoiceDate"].dt.date

# Aggregate total sales by date
daily_sales = (
    df_sales
    .groupby("Date")["TotalSales"]
    .sum()
    .reset_index()
)

# Convert Date back to datetime
daily_sales["Date"] = pd.to_datetime(
    daily_sales["Date"]
)

# Sort by date
daily_sales = daily_sales.sort_values("Date")

# Plot sales trend
plt.figure(figsize=(14, 6))

sns.lineplot(
    data=daily_sales,
    x="Date",
    y="TotalSales"
)

plt.title("Sales Trends Over Time")
plt.xlabel("Date")
plt.ylabel("Total Sales")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 7. HISTOGRAM TO VISUALIZE NUMBER OF INVOICES
# ============================================================

# Count unique invoices for each day
invoices_per_day = (
    df_sales
    .groupby("Date")["InvoiceNo"]
    .nunique()
    .reset_index(name="NumberOfInvoices")
)

# Plot histogram
plt.figure(figsize=(10, 6))

sns.histplot(
    data=invoices_per_day,
    x="NumberOfInvoices",
    bins=30,
    kde=True
)

plt.title("Distribution of Number of Invoices per Day")
plt.xlabel("Number of Invoices per Day")
plt.ylabel("Number of Days")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 8. HISTOGRAM TO VISUALIZE NEGATIVE QUANTITIES
# ============================================================

# Use the original dataset to identify negative quantities
# because df_sales contains only positive quantities.

df_original = pd.read_csv(
    "data.csv",
    encoding="latin1"
)

# Select negative quantities
negative_quantity = df_original[
    df_original["Quantity"] < 0
]

# Display number of negative quantities
print(
    "Number of negative quantity transactions:",
    len(negative_quantity)
)

# Plot histogram
plt.figure(figsize=(10, 6))

sns.histplot(
    data=negative_quantity,
    x="Quantity",
    bins=30
)

plt.title("Distribution of Negative Quantities")
plt.xlabel("Negative Quantity")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 9. TOTAL PER INVOICE
#    Quantity * UnitPrice
#    HISTOGRAM + KDE
# ============================================================

# Calculate total value of each transaction
df_sales["InvoiceLineTotal"] = (
    df_sales["Quantity"] *
    df_sales["UnitPrice"]
)

# Aggregate transaction values by invoice
invoice_total = (
    df_sales
    .groupby("InvoiceNo")["InvoiceLineTotal"]
    .sum()
    .reset_index(name="InvoiceTotal")
)

# Plot invoice total distribution
plt.figure(figsize=(10, 6))

sns.histplot(
    data=invoice_total,
    x="InvoiceTotal",
    bins=50,
    kde=True
)

plt.title("Distribution of Total Sales per Invoice")
plt.xlabel("Invoice Total")
plt.ylabel("Number of Invoices")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 10. DISTRIBUTION OF Quantity
# ============================================================

plt.figure(figsize=(10, 6))

# KDE plot for Quantity
sns.kdeplot(
    data=df_sales,
    x="Quantity",
    fill=True
)

plt.title("Distribution of Quantity")
plt.xlabel("Quantity")
plt.ylabel("Density")

plt.tight_layout()
plt.show()
```

```python

# ============================================================
# 11. DISTRIBUTION OF UNIQUE PRODUCTS SOLD
#     StockCode
# ============================================================

# Count the number of unique products
unique_products = df_sales["StockCode"].nunique()

print(
    "Number of unique products:",
    unique_products
)

# Count how many transactions each StockCode appears in
product_frequency = (
    df_sales
    .groupby("StockCode")
    .size()
    .reset_index(name="TransactionCount")
)

# Plot distribution of product transaction frequency
plt.figure(figsize=(10, 6))

sns.histplot(
    data=product_frequency,
    x="TransactionCount",
    bins=50,
    kde=True
)

plt.title("Distribution of Unique Products Sold")
plt.xlabel("Number of Transactions per StockCode")
plt.ylabel("Number of Products")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 12. SALES PER COUNTRY BY NUMBER OF TRANSACTIONS
# ============================================================

# Count unique invoices for each country
country_transactions = (
    df_sales
    .groupby("Country")["InvoiceNo"]
    .nunique()
    .sort_values(ascending=False)
)

# Plot number of transactions by country
plt.figure(figsize=(14, 8))

sns.barplot(
    x=country_transactions.values,
    y=country_transactions.index
)

plt.title("Number of Transactions per Country")
plt.xlabel("Number of Transactions")
plt.ylabel("Country")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 13. DISTINCT CUSTOMERS PER COUNTRY
# ============================================================

# Remove missing CustomerIDs
customer_country = df_sales.dropna(
    subset=["CustomerID"]
)

# Count distinct customers by country
customers_per_country = (
    customer_country
    .groupby("Country")["CustomerID"]
    .nunique()
    .sort_values(ascending=False)
)

# Plot distinct customers per country
plt.figure(figsize=(14, 8))

sns.barplot(
    x=customers_per_country.values,
    y=customers_per_country.index
)

plt.title("Number of Distinct Customers per Country")
plt.xlabel("Number of Distinct Customers")
plt.ylabel("Country")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 14. DISTRIBUTION OF ITEMS SOLD PER StockCode
# ============================================================

# Calculate total quantity sold for each StockCode
items_per_stockcode = (
    df_sales
    .groupby("StockCode")["Quantity"]
    .sum()
    .reset_index(name="ItemsSold")
)

# Plot distribution
plt.figure(figsize=(10, 6))

sns.histplot(
    data=items_per_stockcode,
    x="ItemsSold",
    bins=50,
    kde=True
)

plt.title("Distribution of Items Sold per StockCode")
plt.xlabel("Total Items Sold")
plt.ylabel("Number of StockCodes")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# 15. UnitPrice AFTER ROUNDING TO NEAREST INTEGER
# ============================================================

# Round UnitPrice to the nearest integer
df_sales["RoundedUnitPrice"] = (
    df_sales["UnitPrice"].round()
)

# Plot distribution
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df_sales,
    x="RoundedUnitPrice",
    bins=50,
    kde=True
)

plt.title("Distribution of Unit Price After Rounding")
plt.xlabel("Rounded Unit Price")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```
