```python
import matplotlib.pyplot as plt
```

```python
# 4. Use relplot() for Combined Plotting -
# 5. Create a line plot of average scores (math, reading, writing) vs parental education level. Hint: use groupby.

# Average scores by parental education
avg_scores = (
    student_df.groupby("parental level of education")[['math score','reading score','writing score']]
    .mean()
    .reset_index()
)

# method 1 : using 3 different lineplot()s

# Line plot
sns.lineplot(
    data=avg_scores,
    x='parental level of education',
    y='math score'
)

sns.lineplot(
    data=avg_scores,
    x='parental level of education',
    y='reading score'
)

sns.lineplot(
    data=avg_scores,
    x='parental level of education',
    y='writing score'
)

plt.title("Average Scores vs Parental Education")
plt.xlabel("parental level of education")
plt.ylabel("Average Score")
plt.xticks(rotation = 45)
plt.show()
```

```python
# method 2 : avoid using 3 different lineplot()s  

# Convert to long format for plotting  - avg_scores_long - used for multiple columns
avg_scores_long = avg_scores.melt(
    id_vars='parental level of education',     # id_vars - columns to be kept, i.e. unchanged
    var_name='Subject',                        # var_name - name for the old column names       
    value_name='Average Score'                 # value_name - name for the old values
)

# Line plot
sns.lineplot(
    data=avg_scores_long,
    x='parental level of education',
    y='Average Score',
    hue='Subject',
    markers='o'
)

plt.title("Average Scores vs Parental Education")
plt.xlabel("parental level of education")
plt.ylabel("Average Score")
plt.xticks(rotation = 45)
plt.show()
```

```python
# 6. Compare Math Scores by Gender Across Parental Education - Overlay gender differences on math scores sorted by parental education.

# Average math score by parental education and gender
math_gender = (student_df.groupby(['parental level of education', 'gender'])['math score'].mean().reset_index())

# Define desired education order
education_order = [
    "some high school",
    "high school",
    "some college",
    "associate's degree",
    "bachelor's degree",
    "master's degree"
]

# Convert education to ordered categorical
math_gender['parental level of education'] = pd.Categorical(
    math_gender['parental level of education'],
    categories=education_order,
    ordered=True
)

# Sort
math_gender = math_gender.sort_values(
    'parental level of education'
)

# Plot
sns.lineplot(
    data=math_gender,
    x='parental level of education',
    y='math score',
    hue='gender',
    marker='o'
)

plt.title('Average Math Scores by Gender and Parental Education')
plt.xlabel('Parental Level of Education')
plt.ylabel('Average Math Score')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# 7. Trend of Average Total Score via relplot() - Create a relational line plot of average total score (mean of three scores) 
# per parental education.

# Calculate total score for each student
student_df['total score'] = (
    student_df['math score'] +
    student_df['reading score'] +
    student_df['writing score']
)

# Average total score by parental education
avg_total = (
    student_df.groupby('parental level of education')['total score']
      .mean()
      .reset_index()
)

# Set education order
avg_total['parental level of education'] = pd.Categorical(
    avg_total['parental level of education'],
    categories=education_order,
    ordered=True
)

print(avg_total)

# Sort by education level
avg_total = avg_total.sort_values(
    'parental level of education'
)

print(avg_total)

# Relational line plot
sns.relplot(
    data=avg_total,
    x='parental level of education',
    y='total score',
    kind='line',
    marker='o',
    height=6,
    aspect=1.5
)

plt.title('Average Total Score by Parental Education')
plt.xlabel('Parental Level of Education')
plt.ylabel('Average Total Score')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

**CATEGORICAL PLOT**

* it deals with categorical variables
* used for visualizing the relationship between variables.
* Those variables can be either be completely numerical or a category like a group, class or division.

**Functions for Categorical Plots in Seaborn**

1. **boxplot()** - Shows distribution (median, quartiles, outliers).the dots indicating the presence of outliers.

syntax :
sns.boxplot([x, y, hue, data, order, hue_order, …])


2. **violinplot()** - Combines boxplot + KDE ( kernal density estimation).estimate the probability distribution of a dataset (continuous variable).

syntax:
sns.violinplot([x, y, hue, data, order, …])

3. **barplot()** - shows mean value of numerical variable for each category (with error bars).

syntax :
sns.barplot([x, y, hue, data, order, hue_order, …])

4. **countplot()**- counts the categories and returns a count of their occurrences. Shows frequency of each category.

syntax:
sns.countplot([x, y, hue, data, order, …])

5. **Stripplo**t -creates a scatter plot based on  a category

syntax:
sns.stripplot([x, y, hue, data, order, …])

6. **swarmplot()** - Like stripplot, but adjusts points to avoid overlap.

syntax:
sns.swarmplot([x, y, hue, data, order, …])

7. **catplot()** - General function for categorical plots.kind can be: "strip", "swarm", "box", "violin", "boxen", "point", "bar", "count".

syntax:
sns.catplot([x, y , data , kind,...])

8. **pointplot()** - Shows mean value with lines connecting categories (good for trends).

syntax :
sns.pointplot([ x, y. data,..])

**KERNAL DENSITY ESTIMATE (KDE)**

 KDE is a way to estimate the probability density function (PDF) of a continuous variable.

 It can be used for both univariate and bivariate distributions, making it a versatile option for exploring data.

 * Univariate KDE: Visualizes the distribution of a single variable.

* Bivariate KDE: Displays the joint distribution of two variables.

**DISTRIBUTION PLOT**

* It provides a high-level interface for drawing attractive and informative statistical graphics.
* used for examining univariate and bivariate distributions.
* A distribution plot shows how data values are spread (their frequency, density, or probability) across a range.

**Common Types of Distribution Plots**

* **Histogram**  → Bars showing frequency counts.

syntax

sns.histplot(data=df, x="numeric_column", bins=20, kde=True)

* **Kernel Density Estimate (KDE)**  → Smooth curve estimate of probability density.

syntax

sns.kdeplot(data=df, x="numeric_column", hue="category_column", fill=True)


* **Combination (Hist + KDE)** → Both histogram & KDE together.


```python
# example for univariate  - Analysis of one variable at a time.

import seaborn as sns
import matplotlib.pyplot as plt

# Example data
data = [1, 2, 2.5, 3, 3.5, 4, 5]

# Univariate KDE
sns.kdeplot(data=data, fill=True, color="blue")
plt.show()
```

```python
# example for bivariate - Analysis of the relationship between two variables.
import seaborn as sns
import matplotlib.pyplot as plt

# Example data
import numpy as np
x = np.random.normal(size=100)
y = np.random.normal(size=100)

# Bivariate KDE
sns.kdeplot(x=x, y=y, cmap="Blues", fill=True)
plt.show()
```

```python
# example
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
iris = sns.load_dataset("iris")
iris
```

```python
# Histogram - Shows frequency counts of values.
sns.histplot(iris['petal_length'], bins=20, color="skyblue")
plt.title("Histogram of Petal Length")
plt.show()
```

**REGESSION PLOT** - show data points and the best-fit line that represents the relationship between variables.

REGRESSION  

 statistical method used to understand the relationship between variables. predict the value of dependent variable based on the values of independent variables.

* **regplot**() -> Scatter plot with regression line


syntax

sns.regplot(

    data=None,       # Your DataFrame
    x=None,          # Column name for x-axis (numeric)
    y=None,          # Column name for y-axis (numeric)
    ci=95,           # Confidence interval for regression line (default 95%)
    scatter=True,    # Show scatter points (True/False)
    fit_reg=True,    # Show regression line (True/False)
    order=1,         # Polynomial order (1=linear, 2=quadratic, etc.)
    logx=False,      # Apply log scale on x-axis (True/False)
    color=None       # Color of points and line
)

**log scale**

* The axis increases by powers of 10 (1, 10, 100, 1000,
* This means small values are stretched apart and large values are compressed.

**Why Use Log Scale?**

* When your data spans a very large range (e.g., from 1 to 10000).

* To make patterns easier to see in skewed data.

* To handle exponential relationships between variables.

**polynomial** - Type of regression , modeled as a polynomial (curved) function rather than a straight line.

**Why Use Polynomial Regression?**

* When the relationship between variables is not linear.
* Captures curved trends in data.

 **Confidence Interval (CI)**

A confidence interval is a range of values that is likely to contain the true value of a parameter (like a regression line) with a certain level of confidence (usually 95%).

* If you collect the same data many times and fit regression lines, most of those lines would fall within the shaded confidence interval.

* For a 95% CI, we are 95% confident the true regression line lies inside that shaded region.

**Why Important?**

* Shows uncertainty around the regression line.

* Wide CI → predictions are less reliable.

* Narrow CI → predictions are more reliable.


 **lmplot()**: Combines regression and faceting for multiple subsets

syntax:

sns.lmplot(

    data=None,       # Your DataFrame
    x=None,          # Column for x-axis
    y=None,          # Column for y-axis
    hue=None,        # Categorical variable to color by
    col=None,        # Column to facet by
    row=None,        # Row to facet by
    ci=95,           # Confidence interval
    scatter=True,    # Show scatter points
    order=1,         # Polynomial regression order
    logx=False,      # Log scale for x-axis
    palette=None     # Color palette for hue
)




```python
# Basic Regression Plot
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
tips = sns.load_dataset("tips")

sns.regplot(x="total_bill", y="tip", data=tips)

plt.title("Basic Regression Plot")
plt.show()
```

```python
# Regression Line Only (No Scatter)
sns.regplot(x="total_bill", y="tip", data=tips, scatter=False, color="red")
plt.title("Regression Line Only")
plt.show()
```

```python
# Polynomial Regression (Order 2)
sns.regplot(x="total_bill", y="tip", data=tips, order=2, line_kws={"color":"green"})
plt.title("Polynomial Regression (Quadratic Fit)")
plt.show()
```

```python
# Confidence Interval Disabled
sns.regplot(x="total_bill", y="tip", data=tips, ci=95)

plt.title("Regression Plot without Confidence Interval")
plt.show()
```

```python
# Multiple Categories using lmplot
sns.lmplot(x="total_bill", y="tip", hue="smoker", data=tips, height=6, aspect=1.3)
plt.title("Regression Plot Grouped by Smoker/Non-Smoker")
plt.show()
```

**correlation** - Correlation is a statistical measure that tells us how two variables are related to each other.

**Types of Correlation**

* Positive Correlation - When one variable increases, the other also increases.
* Negative Correlation - When one variable increases, the other decreases.
* No Correlation - Variables do not show any clear relationship.

**Why Correlation is Useful**

* Helps to find relationships between variables before building models.
* Helps identify which variables move together or are independent.
* Used in heatmaps, feature selection, and clustering.

**Correlation Coefficient**

A correlation coefficient is a numerical value that measures the strength and direction of a relationship between two variables.

Denoted by r, ranges from -1 to 1.

r value	Meaning
1	Perfect positive correlation
-1	Perfect negative correlation
0	No correlation

To calculate correlation :

corr = dataframe[column].corr()




**Color Map Name	Description / Usage**
* 'coolwarm'	  Blue = negative, Red = positive
* 'viridis'	    Green → Yellow, good for sequential data
* 'plasma'	    Dark → Bright, sequential
* 'YlGnBu'    	Yellow → Green → Blue, sequential
* 'Reds'	      Shades of red, good for intensity
* 'Blues'       	Shades of blue, good for correlations
* 'RdBu_r'     	Red ↔ Blue diverging, centered around 0
* 'magma'	    Dark → Light, sequential

```python
# Heatmap
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# Create random matrix
data = np.random.rand(6, 6)

# Heatmap
sns.heatmap(data, annot=True, cmap="YlGnBu")

plt.title("Heatmap Example")
plt.show()

# annot=True writes numbers on the heatmap.
# cmap controls color (examples: "coolwarm", "viridis", "YlGnBu").
```

```python
# Correlation Heatmap (Real Dataset)
# Load dataset
tips = sns.load_dataset("tips")

# Select only numerical columns
numeric_tips = tips.select_dtypes(include=['number'])

# Correlation matrix
corr = numeric_tips.corr()

# Heatmap of correlation
sns.heatmap(corr, annot=True, cmap="coolwarm", linewidths=0.5)

plt.title("Correlation Heatmap of Tips Dataset")
plt.show()
```

```python
# Masked Heatmap (Half Matrix)
mask = np.triu(np.ones_like(corr, dtype=bool))  # Upper triangle mask

sns.heatmap(corr, annot=True, cmap="coolwarm", mask=mask)

plt.title("Masked Heatmap (Lower Triangle Only)")
plt.show()

# Helps when correlation matrix is symmetric.
```

```python
# Clustermap
sns.clustermap(corr, annot=True, cmap="viridis")

plt.title("Cluster Map of Correlations")
plt.show()

# Automatically groups similar rows and columns together.
```

**DATA MERGING AND JOINING**

* In real-world datasets, information may be spread across multiple tables (DataFrames).

* To analyze properly, we need to combine them into one DataFrame.

* Pandas provides tools for this:

* Method	Use Case

merge()	Combine two DataFrames on a common column (like SQL joins).

concat()	Stack two DataFrames vertically or horizontally.

join()	Similar to merge, but works with index instead of columns.

* Once merged, we can easily use Seaborn for visualization.

**Types of Merge (like SQL joins)**

Inner Join → Keep only matching rows.

Left Join → Keep all rows from the left DataFrame.

Right Join → Keep all rows from the right DataFrame.

Outer Join → Keep all rows from both DataFrames.

image.png

```python
# Merge average tip by day with average total bill

import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

avg_tip = tips.groupby("day")["tip"].mean().reset_index()
avg_bill = tips.groupby("day")["total_bill"].mean().reset_index()

merged = pd.merge(avg_tip, avg_bill, on="day")

sns.barplot(data=merged, x="day", y="tip", color="blue", label="Avg Tip")
sns.barplot(data=merged, x="day", y="total_bill", color="orange", alpha=0.5, label="Avg Bill")
plt.legend()
plt.title("Average Tip and Total Bill by Day")
plt.show()

# Compares average tips vs bills for each day.
```
