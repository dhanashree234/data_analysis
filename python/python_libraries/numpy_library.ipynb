1. Find standard deviation and variance.
2. For arr2D = np.array([[1,2,3],[4,5,6]]), find row-wise and column-wise sum.
3. Square and square root of array [1,4,9].
4. Compute exponential of [1,2,3].
5. Compute sin, cos, tan of [0, π/2, π].
6. Generate 2×4 random integers between 0–100.
7. Set seed 42 and generate 5 random numbers.
8. Create 1D array of 10 random numbers from normal distribution.
9. Generate 7 equally spaced numbers from 2 to 14.
10. Generate 4 numbers between 0–10.
11. Compare 5 points between 0–1 using linspace vs arange.
12. Create numbers from 5–20 with step 3.
13. Create numbers from 10–1 in reverse order.
14. Create 10 numbers from 0–9 using arange.
15. Compare arange(0,2,0.3) with linspace(0,2,7).

```python
import numpy as np

# 1. Find standard deviation and variance
arr = np.array([1, 2, 3, 4, 5])

print("1. Standard Deviation:", np.std(arr))
print("   Variance:", np.var(arr))
```

```python
# 2. Row-wise and column-wise sum of 2D array
arr2D = np.array([[1, 2, 3],
                  [4, 5, 6]])

print("\n2. Row-wise sum:", np.sum(arr2D, axis=1))
print("   Column-wise sum:", np.sum(arr2D, axis=0))
```

```python
# 3. Square and square root of array
arr = np.array([1, 4, 9])

print("\n3. Square:", np.square(arr))
print("   Square Root:", np.sqrt(arr))
```

```python
# 4. Exponential of [1,2,3]
arr = np.array([1, 2, 3])

print("\n4. Exponential:", np.exp(arr))
```

```python
# 5. Sin, Cos, Tan of [0, π/2, π]
arr = np.array([0, np.pi/2, np.pi])

print("\n5. Sin:", np.sin(arr))
print("   Cos:", np.cos(arr))
print("   Tan:", np.tan(arr))
```

```python
# 6. Generate 2x4 random integers between 0–100
random_int = np.random.randint(0, 101, size=(2, 4))

print("\n6. Random 2x4 integers:")
print(random_int)
```

```python
# 7. Set seed 42 and generate 5 random numbers
np.random.seed(42)

random_numbers = np.random.rand(5)

print("\n7. Random numbers with seed 42:")
print(random_numbers)
```

```python
# 8. Create 1D array of 10 random numbers from normal distribution
normal_array = np.random.randn(10)

print("\n8. Normal distribution random numbers:")
print(normal_array)
```

```python
# 9. Generate 7 equally spaced numbers from 2 to 14
arr = np.linspace(2, 14, 7)

print("\n9. 7 equally spaced numbers:")
print(arr)
```

```python
# 10. Generate 4 numbers between 0–10
arr = np.linspace(0, 10, 4)

print("\n10. 4 numbers between 0 and 10:")
print(arr)
```

```python
# 11. Compare 5 points between 0–1 using linspace vs arange
lin = np.linspace(0, 1, 5)
ar = np.arange(0, 1.25, 0.25)
arr = np.arange(0, 1, 5)

print("\n Linspace:")
print(lin)

print("    Arange:")
print(arr)

print("    Arange:")
print(ar)
```

```python
# 12. Create numbers from 5–20 with step 3
arr = np.arange(5, 21, 3)

print("\n12. Numbers from 5 to 20 step 3:")
print(arr)
```

```python
# 13. Create numbers from 10–1 in reverse order
arr = np.arange(10, 0, -1)

print("\n13. Reverse order:")
print(arr)
```

```python
# 14. Create 10 numbers from 0–9 using arange
arr = np.arange(10)

print("\n14. Numbers from 0 to 9:")
print(arr)
```

```python
# 15. Compare arange(0,2,0.3) with linspace(0,2,7)
arange_arr = np.arange(0, 2, 0.3)
linspace_arr = np.linspace(0, 2, 7)

print("\n15. arange(0,2,0.3):")
print(arange_arr)

print("    linspace(0,2,7):")
print(linspace_arr)
```
