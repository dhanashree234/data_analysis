Exception Handling

```python
# try
#            |
#       Is there error?
#         /         \
#       Yes          No
#       |             |
#  except        Remaining Code
```

```python
# SYNTAX: 
# try:
#     # Risky Code
# except:
#     # Runs if error occurs
```

```python
try:
    num = 10 / 0
except:
    print("Cannot divide by zero")
```

```python
try:
    number = int(input("Enter a number: "))
    print(100 / number)
except:
    print("Something went wrong")
```

```python
try:
    a = int(input("Enter number: "))
    print(100 / a)

except ZeroDivisionError:
    print("Division by zero is not allowed.")
```

```python
try:
    age = int(input("Enter age: "))
    print(age)

except ValueError:
    print("Enter numbers only.")
```

```python
try:
    a = int(input("Enter number: "))
    print(100 / a)

except ValueError:
    print("Enter numbers only.")

except ZeroDivisionError:
    print("ZeroDivisionError")

```

```python
try:
    x = int(input("Enter number: "))
    print(100 / x)

except Exception as e:
    print(e)
```

```python
try:
    num = int(input("Enter number: "))
    print(100 / num)

except ZeroDivisionError:
    print("Cannot divide by zero")

else:
    print("Program executed successfully")
```

```python
try:
    num = int(input("Enter number: "))
    print(100 / num)

except:
    print("Error")

finally:
    print("Program Ended")
```

```python
age = int(input("Enter age: "))

try:
    if age < 18:
        raise Exception("You are not eligible.")

    print("Eligible")

except Exception as e:
    print(e)
```

```python
balance = 5000

try:
    amount = int(input("Enter withdrawal amount: "))

    if amount > balance:
        raise Exception("Insufficient Balance")

    balance -= amount

    print("Remaining Balance:", balance)

except Exception as e:
    print(e)
```

```python
username = input("Username: ")
password = input("Password: ")

try:
    if username != "admin" or password != "1234":
        raise Exception("Invalid Login")

    print("Login Successful")

except Exception as e:
    print(e)
```

```python
try:
    file = open("q.txt", "r")
    print(file.read())
    file.close()
except FileNotFoundError:
    print("file not found error") 
```

ENUMERATE function    
enumerate(iterable, start=0) - retrieve data value or index

```python
fruits = ["Apple", "Banana", "Orange"]

for fruit in fruits:
    print(fruit)
```

```python
fruits = ["Apple", "Banana", "Orange"]

for index, fruit in enumerate(fruits, start=1):
    print(index, fruit)
```

```python
fruits = ["Apple", "Banana", "Orange"]

for index, fruit in enumerate(fruits):
    if fruit == "Banana":
        print("Found at", index)
```

```python
fruits = ["Apple", "Banana", "Orange"]

result = list(enumerate(fruits))

print(result) # collection of tuple in list
```

```python
colors = ("Red", "Green", "Blue")

for i, color in enumerate(colors, start=1):
    print(i, color)
```

```python
name = "python"

for i, letter in enumerate(name):
    print(i,letter)
```

```python
students = ["qwe","asd","zxc"]

for rno, name in enumerate(students, start=1):
    print(f"Roll no: {rno}  Name : {name}")
```

```python
marks = [10,20,10,20]

for i, mark in enumerate(marks):
    print("Student", i+1, " marks:", mark)
```

```python
# Print employee names with IDs starting from 101.

employees = ["qwe","asd","zxc"]

for emp_id, name in enumerate(employees, start=101):
    print(f"emp_id: {emp_id}  Name : {name}")
```

```python
numbers = [1,2,3,4,5]

largest = max(numbers)

for i, num in enumerate(numbers):
    if num == largest:
        print("Largest number is ", num)
        print("Index is ", i)
```

```python
# print the even index with value
names = ["name1", "name2", "name3"]

for i, n in enumerate(names):
    if i % 2 == 0:
        print(i,n)
```
