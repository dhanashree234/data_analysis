```python
# to check if the file is empty
```

```python
filename = "1.txt"

with open(filename, "r") as file:
    content = file.read()

if content == "":
    print("File is empty")
else:
    print("File is not empty")
```

```python
import os

filename = "1.txt"

if os.path.getsize(filename) == 0:
    print("File is empty")
else:
    print("File is not empty")
```

```python
# 2. Create a file and write five favorite programming languages into it.
```

```python
with open("languages.txt", "w") as file:
    file.write("Python\n")
    file.write("Java\n")
    file.write("C\n")
    file.write("C++\n")
    file.write("JavaScript\n")

print("File written")
```

```python
# 3. Read and display the entire contents of a file.

with open("languages.txt", "r") as file:
    print(file.read())
```

```python
# 4. Read only the first 10 characters from a file.
with open("languages.txt", "r") as file:
    print(file.read(10))
```

```python
# 5. Read one line at a time using readline().
with open("languages.txt", "r") as file:
    print(file.readline())
    print(file.readline())
    print(file.readline())
    print(file.readline())
```

```python
# 6. Read all lines using readlines().
with open("languages.txt", "r") as file:
    lines = file.readlines()
    
print(lines)
```

```python
# 7. Count the total number of lines in a file.
print(len(lines))
```

```python
# 8. Count the total number of words in a file.
with open("languages.txt", "r") as file:
    text = file.read()

words = text.split()
print("Total words:", len(words))
```

```python
# 9. Count the total number of characters in a file.
with open("languages.txt", "r") as file:
    text = file.read()

print("Total characters:", len(text))
```

```python
# 10. Display the current file pointer position using tell().
with open("languages.txt", "r") as file:
    print("Current position:", file.tell())
```

```python
# 11. Read 5 characters from a file and display the file pointer position using tell().
with open("languages.txt", "r") as file:
    print(file.read(5))
    print("Pointer position:", file.tell())
```

```python
# 12. Move the file pointer to the beginning using seek() and read the file again.
with open("languages.txt", "r") as file:
    file.seek(0)
    print(file.read())
```

```python
# 13. Close a file properly using close().
file = open("languages.txt", "r")
print(file.read())
file.close()

print("File closed")    
```

```python
# 14. Check whether a file exists before opening it.
import os

filename = "languages.txt"

if os.path.exists(filename):
    with open(filename, "r") as file:
        print(file.read())
else:
    print("File does not exist")
```

```python
# 15. Create a file only if it does not already exist.
try:
    with open("newfile.txt", "x") as file:
        file.write("This file is created only once.")

    print("File created")
except FileExistsError:
    print("File already exists")
```
