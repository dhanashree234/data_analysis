11.	Write a Python program to take the student's age and marks. Check whether the student is eligible if the age is at least 18 and the marks are at least 50.

```python
class Eligibility:
    def accept(self):
        self.age = int(input("Enter age: "))
        self.marks = float(input("Enter marks: "))

    def check(self):
        return self.age >= 18 and self.marks >= 50

    def display(self):
        print("Eligible:", self.check())
```

```python
obj = Eligibility()
```

```python
obj.accept()
```

```python
obj.display()
```
