13.	A customer gets a discount if their age is at least 60 or they are a student. Write a program to check discount eligibility.

```python
class Discount:
    def accept(self):
        self.age = int(input("Enter age: "))
        self.student = input("Are you a student? (yes/no): ").strip().lower()

    def check(self):
        return self.age >= 60 or self.student == "yes"

    def display(self):
        print("Eligible for Discount:", self.check())
```

```python
obj = Discount()
```

```python
obj.accept()
```

```python
obj.display()
```
