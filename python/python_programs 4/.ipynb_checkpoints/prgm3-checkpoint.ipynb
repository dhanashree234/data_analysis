3.	Write a Python program to take two numbers and perform normal division and floor division.

```python
class Division:
    def accept(self):
        self.num1 = float(input("Enter first number: "))
        self.num2 = float(input("Enter second number: "))

    def calculate(self):
        return self.num1 / self.num2, self.num1 // self.num2

    def display(self):
        division, floor_division = self.calculate()
        print("Normal Division =", division)
        print("Floor Division =", floor_division)
```

```python
obj = Division()
```

```python
obj.accept()
```

```python
obj.display()
```
