4.	Write a Python program to store a number and increase it by 10 using the += operator.

```python
class Increment:
    def accept(self):
        self.num = int(input("Enter a number: "))

    def calculate(self):
        self.num += 10
        return self.num

    def display(self):
        print("Updated Number =", self.calculate())
```

```python
obj = Increment()
```

```python
obj.accept()
```

```python
obj.display()
```
