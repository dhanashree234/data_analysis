1. Write a Python program to take two numbers and find the remainder after division.

```python
class Remainder:
    def accept(self):
        self.num1 = int(input("Enter first number: "))
        self.num2 = int(input("Enter second number: "))

    def calculate(self):
        return self.num1 % self.num2

    def display(self):
        print("Number 1 and 2 =", self.num1, "and", self.num2)
        print("Remainder =", self.calculate())
```

```python
obj = Remainder()
```

```python
obj.accept()
```

```python
obj.display()
```
