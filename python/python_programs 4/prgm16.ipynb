16.	A training centre provides "Python", "Java", and "Web Development" courses. Take the student's age and selected course. Check whether the selected course exists and whether the student is at least 16 years old.

```python
class TrainingCentre:
    def accept(self):
        self.age = int(input("Enter age: "))
        self.course = input("Enter course: ").strip().title()

    def check(self):
        courses = ["Python", "Java", "Web Development"]
        course_exists = self.course in courses
        eligible = self.age >= 16
        return course_exists, eligible

    def display(self):
        course_exists, eligible = self.check()

        print("Course Available:", course_exists)
        print("Age Eligible:", eligible)

        if course_exists and eligible:
            print("Admission Approved")
        else:
            print("Admission Not Approved")
```

```python
obj = TrainingCentre()
```

```python
obj.accept()
```

```python
obj.display()
```
