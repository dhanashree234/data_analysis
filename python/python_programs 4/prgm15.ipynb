15.	A student qualifies for an exam if their attendance is at least 75% and their assignment mark is at least 50. Write a Python program to check qualification.

```python
class Exam:
    def accept(self):
        self.attendance = float(input("Enter attendance (%): "))
        self.assignment = float(input("Enter assignment marks: "))

    def check(self):
        return self.attendance >= 75 and self.assignment >= 50

    def display(self):
        print("Qualified:", self.check())
```

```python
obj = Exam()
```

```python
obj.accept()
```

```python
obj.display()
```
