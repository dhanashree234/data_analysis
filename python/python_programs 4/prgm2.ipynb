2.	Write a Python program to take a number and calculate its square and cube.

```python
class SquareCube:
    def accept(self):
        self.num = int(input("Enter a number: "))

    def calculate(self):
        return self.num ** 2, self.num ** 3

    def display(self):
        square, cube = self.calculate()
        print("Square =", square)
        print("Cube =", cube)
```

```python
obj = SquareCube()
```

```python
obj.accept()
```

```python
obj.display()
```
