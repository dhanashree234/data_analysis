10.	Write a program to take a product price and check whether the price is less than £100.

```python
class PriceCheck:
    def accept(self):
        self.price = float(input("Enter product price: £"))

    def check(self):
        return self.price < 100

    def display(self):
        print("Price is less than £100:", self.check())
```

```python
obj = PriceCheck()
```

```python
obj.accept()
```

```python
obj.display()
```
