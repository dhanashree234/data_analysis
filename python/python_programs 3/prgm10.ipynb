10. Create a simple employee salary program that accepts basic salary and bonus and displays the total salary.

Create class and methods

```python
class Employee:
    def accept(self):
        self.salary = int(input("Enter salary:"))
        self.bonus = int(input("Enter bonus:"))
        
    def calculate(self):
        return self.salary + self.bonus

    def display(self):
        total = self.calculate()
        print(f"Conversion = {total}")
```

Create object

```python
n_obj = Employee()
```

```python
n_obj.accept()
```

```python
n_obj.display()
```
