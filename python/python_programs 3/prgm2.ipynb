2. Take two numbers and display their sum.

Create class and methods

```python
class Number:
    def accept(self):
        self.num1 = int(input("Enter first number:"))
        self.num2 = int(input("Enter second number:"))

    def calculate(self):
        return self.num1 + self.num2

    def display(self):
        result = self.calculate()
        print(f"Result = {result}")
```

```python
# Note : In Python, self is a reference to the current object (instance) of a class. 
# It allows you to access the object's variables (attributes) and methods.
# Why do we use self?
# To store data that belongs to a specific object.
# To access the object's attributes and methods inside the class.
```

Create object

```python
n_obj = Number()
```

```python
n_obj.accept()
```

```python
n_obj.display()
```
