6. Take marks for five subjects and calculate total and average.

Create class and methods

```python
class StudentResult:
    def accept(self):
        self.name = input("Enter name:")
        self.marks1 = int(input("Enter marks for subject1:"))
        self.marks2 = int(input("Enter marks for subject2:"))
        self.marks3 = int(input("Enter marks for subject3:"))
        self.marks4 = int(input("Enter marks for subject4:"))
        self.marks5 = int(input("Enter marks for subject5:"))
        
    def calculate(self):
        total = self.marks1 + self.marks2 + self.marks3 + self.marks4 + self.marks5
        average = total / 5
        return total, average

    def display(self):
        total, average = self.calculate()
        print(f"Total = {total} and average = {average}")
```

Create object

```python
n_obj = StudentResult()
```

```python
n_obj.accept()
```

```python
n_obj.display()
```
