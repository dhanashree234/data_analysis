3. Take two numbers and calculate addition, subtraction, multiplication, and division.

Create class and methods

```python
class Operations:
    def accept(self):
        self.num1 = int(input("Enter first number:"))
        self.num2 = int(input("Enter second number:"))

    def calculate_sum(self):
        return self.num1 + self.num2

    def calculate_diff(self):
        return self.num1 - self.num2

    def calculate_mul(self):
        return self.num1 * self.num2

    def calculate_div(self):
        return self.num1 / self.num2

    def display(self):
        print(f"First number = {self.num1} \tSecond number = {self.num2}")

    def display_sum(self):
        result = self.calculate_sum()
        print(f"Sum = {result}")

    def display_diff(self):
        result = self.calculate_diff()
        print(f"Difference = {result}")

    def display_mul(self):
        result = self.calculate_mul()
        print(f"Multiplication = {result}")

    def display_div(self):
        result = self.calculate_div()
        print(f"Division = {result}")    
```

```python
# Note : In Python, self is a reference to the current object (instance) of a class. 
# It allows you to access the object's variables (attributes) and methods.
# Why do we use self?
# To store data that belongs to a specific object.
# To access the object's attributes and methods inside the class.
```

Create object

```python
n_obj = Operations()
```

```python
n_obj.accept()
```

```python
n_obj.display()
n_obj.display_sum()
n_obj.display_diff()
n_obj.display_mul()
n_obj.display_div()
```
