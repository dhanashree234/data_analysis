9. Take the user's current age and display their age after 5 years.

Create class and methods

```python
class User:
    def accept(self):
        self.age = int(input("Enter age:"))
        
    def calculate(self):
        return self.age + 5

    def display(self):
        age = self.calculate()
        print(f"Conversion = {age}")
```

Create object

```python
n_obj = User()
```

```python
n_obj.accept()
```

```python
n_obj.display()
```
