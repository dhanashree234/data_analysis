4. Take the length and width of a rectangle and calculate its area and perimeter.

Create class and methods

```python
class Rectangle:
    def accept(self):
        self.length = int(input("Enter length:"))
        self.breadth = int(input("Enter breadth:"))

    def calculate(self):
        return self.length * self.breadth

    def display(self):
        result = self.calculate()
        print(f"Area = {result}")
```

```python
# Note : In Python, self is a reference to the current object (instance) of a class. 
# It allows you to access the object's variables (attributes) and methods.
# Why do we use self?
# To store data that belongs to a specific object.
# To access the object's attributes and methods inside the class.
```

Create object

```python
n_obj = Rectangle()
```

```python
n_obj.accept()
```

```python
n_obj.display()
```
